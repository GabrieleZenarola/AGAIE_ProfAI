1. Chiede all'utente di inserire un numero intero.
2. Verifica se il numero è positivo, negativo o zero.
3. Stampa un messaggio diverso per ciascun caso: "Il numero è positivo", "Il numero è negativo" oppure "Il numero è zero".

Successivamente, modifica il programma in modo che:

3. Se la parola è "python", stampa "Hai inserito la parola segreta!"; altrimenti stampa "Parola non riconosciuta".

Infine, crea un programma che:

2. Stampa un messaggio in base all'età:
   - Se l'età è minore di 18 anni, stampa "Sei minorenne".
   - Se l'età è compresa tra 18 e 65 anni (inclusi), stampa "Sei un adulto".
   - Se l'età è maggiore di 65 anni, stampa "Sei un anziano".

```python
# STEP 1: Scrivo un programma che:
# 1. Chiede all'utente di inserire un numero intero.
# 2. Verifica se il numero è positivo, negativo o zero.
# 3. Stampa un messaggio diverso per ciascun caso: "Il numero è positivo", "Il numero è negativo" oppure "Il numero è zero".

# Richiesta dell'input
number_int = int(input("Inserisci un numero intero: "))

# Verifica
if number_int > 0: # numero intero positivo
    print(f"Il numero {number_int} è positivo")
elif number_int < 0: # numero intero negativo
    print(f"Il numero {number_int} è negativo")
else: # zero
    print("Il numero è zero")

#=============================================================================

# STEP 2: modifico il programma in modo che:
# 1. L'utente inserisca una parola qualsiasi.
# 2. Controlla se la parola è "python" (senza distinzione tra maiuscole e minuscole).
# 3. Se la parola è "python", stampa "Hai inserito la parola segreta!"; altrimenti stampa "Parola non riconosciuta".

# Richiesta dell'input
word_strg = input("Digita una parola: ")

# Verifica
if word_strg.lower() == "python": # trasformo la stringa in minuscolo
    print("Hai inserito la parola segreta!")
else:
    print("Parola non riconosciuta")

#=============================================================================

# STEP 3:  creo un programma che:
# 1. Chiede all'utente di inserire la sua età.
# 2. Stampa un messaggio in base all'età:
  # a. Se l'età è minore di 18 anni, stampa "Sei minorenne".
  # b. Se l'età è compresa tra 18 e 65 anni (inclusi), stampa "Sei un adulto".
  # c. Se l'età è maggiore di 65 anni, stampa "Sei un anziano".

# Richiesta dell'input
age_int = int(input("Inserisci la tua età: "))

# Verifica
if age_int < 18: # caso minorenne
  print("Sei minorenne")
elif age_int >= 18 and age_int <= 65: # caso adulto
  print("Sei un adulto")
else: # caso anziano
  print("Sei un anziano")
```

Utilizza le istruzioni condizionali apprese per controllare le diverse condizioni e stampare il messaggio corrispondente.

#**03-Intermedio**
Scrivi un programma Python che simula un sistema di controllo accessi per un sito web e successivamente valuta un punteggio inserito dall'utente.

Il programma dovrà:
1. Chiedere all'utente di inserire una email.
2. Verificare se l'email corrisponde a quella registrata (esempio: "studente@example.com").
3. Se l'email è corretta, chiedere la password.
4. Controllare che la password sia esatta (esempio: "python2024").
5. Stampare "Accesso consentito" se entrambe le condizioni sono soddisfatte, altrimenti stampare "Accesso negato" con messaggi specifici se l'email è errata o se la password è sbagliata.

Dopo aver gestito l'accesso, il programma chiederà all'utente di inserire un punteggio numerico (da 0 a 100) e stamperà la valutazione corrispondente secondo queste regole:
- Da 90 a 100: "Eccellente"
- Da 70 a 89: "Buono"
- Da 50 a 69: "Sufficiente"
- Meno di 50: "Insufficiente"

Utilizza istruzioni condizionali con if, elif ed else per gestire tutti i casi.

Assicurati di rispettare l'indentazione di Python e di gestire eventuali input errati stampando messaggi chiari.

###**Esempi di Input/Output**
input:
studente@example.com
python2024
85 <br>
output:
Accesso consentito
Buono

input:
user@example.com
anypassword
72 <br>
output:
Email errata. Accesso negato.
Buono

input:
studente@example.com
wrongpass
40 <br>
output:
Password errata. Accesso negato.
Insufficiente

```python
# Definisco la mail e la password valide
valid_email = "studente@example.com"
valid_password = "python2024"

# Definisco una variabile errore per verificare se l'accesso è corretto
correct_access = False

# 1. Chiedo all'utente di inserire una email
user_mail_str = input("Inserisci la tua mail: ")

# 2. Controllo se la mail è corretta
if user_mail_str == valid_email:
  # 3. Chiedo la password
  user_password_str = input("Inserisci la tua password: ")
  # 4. Controllo se la password inserita è valida
  if user_password_str == valid_password:
    # 5. Stampo "Accesso consentito" e imposto la variabile dell'accesso corretto a True
    print("Accesso consentito")
    correct_access = True
  # Gestisco il caso di password errata
  else:
    print("Password errata. Accesso negato.")
# Gestisco il caso di mail errata
else:
  print("Email errata. Accesso negato.")

# Se l'accesso è stato eseguito correttamente, chiedo all'utente di inserire un punteggio numerico (da 0 a 100)
# e stampo la valutazione corrispondente secondo queste regole:
# a. Da 90 a 100: "Eccellente"
# b. Da 70 a 89: "Buono"
# c. Da 50 a 69: "Sufficiente"
# d. Meno di 50: "Insufficiente"
if correct_access:
  # Definisco una variabile errore per gestire l'inserimento di un dato non intero
  score_is_int = True
  # Verifico che l'input richiesto sia effettivamente un nuemero intero
  try:
    user_score_int = int(input("Inserisci il tuo punteggio: "))
  except ValueError:
    print("Il punteggio deve essere un numero intero.")
    score_is_int = False

  # Solo se il dato inserito è intero, verifico che il numero inserito sia compreso tra 0 e 100
  if score_is_int:
    if user_score_int < 0 or user_score_int > 100:
      print("Il punteggio deve essere compreso tra 0 e 100.")
    else:
      # Se il numero inserito è compreso nel rango richiesto, stampo la valutazione corrispondente
      if user_score_int >= 90 and user_score_int <= 100:
        print("Eccellente")
      elif user_score_int >= 70 and user_score_int <= 89:
        print("Buono")
      elif user_score_int >= 50 and user_score_int <= 69:
        print("Sufficiente")
      else:
        print("Insufficiente")
```

#**04-Difficile**
Scrivi un programma Python che simuli un sistema di classificazione avanzata di un numero inserito dall'utente. Il programma deve leggere da tastiera un numero intero e classificare questo numero secondo queste regole, usando esclusivamente istruzioni condizionali (if, elif, else):

1. Se il numero è negativo, stampa "Numero negativo".
2. Se il numero è zero, stampa "Zero".
3. Se il numero è positivo e divisibile per 2 e per 3, stampa "Multiplo di 2 e 3".
4. Se il numero è positivo e divisibile solo per 2, stampa "Multiplo solo di 2".
5. Se il numero è positivo e divisibile solo per 3, stampa "Multiplo solo di 3".
6. Se il numero è positivo ma non è divisibile né per 2 né per 3, stampa "Numero positivo non multiplo di 2 o 3".

Devi utilizzare l'operatore modulo (%) per verificare la divisibilità e strutturare le condizioni in modo corretto sfruttando if, elif e else per evitare controlli ridondanti. Il programma deve essere robusto e mostrare un solo messaggio per ciascun input.

Esempio di esecuzione:
L'utente inserisce un numero e il programma mostra la classificazione corretta.

###**Esempi di Input/Output**
input:
12 <br>
output:
Multiplo di 2 e 3

input:
8 <br>
output:
Multiplo solo di 2

input:
9 <br>
output:
Multiplo solo di 3

input:
7 <br>
output:
Numero positivo non multiplo di 2 o 3

input:
0 <br>
output:
Zero

input:
-5 <br>
output:
Numero negativo

```python
# STEP 1: Richiesta dell'input con controllo errore se non è un intero.
# La richiesta dell'input continua finché l'input è un intero
while True:
  try:
    input_int = int(input("Inserisci un numero intero: "))
    break
  except ValueError:
    print("L'input deve essere un numero intero.")

# STEP 2: Classifico l'input secondo queste regole, usando esclusivamente istruzioni condizionali (if, elif, else):
# a. Se il numero è negativo, stampa "Numero negativo".
if input_int < 0:
  print("Numero negativo")
# b. Se il numero è zero, stampa "Zero".
elif input_int == 0:
  print("Zero")
# c. Se il numero è positivo e divisibile per 2 e per 3, stampa "Multiplo di 2 e 3".
elif input_int % 2 == 0 and input_int % 3 == 0:
  print("Multiplo di 2 e 3")
# d. Se il numero è positivo e divisibile solo per 2, stampa "Multiplo solo di 2".
elif input_int % 2 == 0:
  print("Multiplo solo di 2")
# e. Se il numero è positivo e divisibile solo per 3, stampa "Multiplo solo di 3".
elif input_int % 3 == 0:
  print("Multiplo solo di 3")
# f. Se il numero è positivo ma non è divisibile né per 2 né per 3, stampa "Numero positivo non multiplo di 2 o 3".
else:
  print("Numero positivo non multiplo di 2 o 3")
```

#**05-Esperto**
In questa esercitazione dovrai scrivere un programma che analizza alcune informazioni riguardanti utenti di un sistema e restituisce un messaggio appropriato in base a più condizioni concatenate. Dovrai utilizzare le istruzioni condizionali if, elif ed else e combinare espressioni booleane con operatori logici come AND e OR, sfruttando al meglio l'annidamento e la concatenazione con elif per gestire tutte le casistiche.

Il programma riceverà in input tre dati: il nome utente (stringa), l'età (numero intero), e il livello di accesso (stringa, uno tra: 'guest', 'user', 'admin'). Il programma dovrà eseguire i seguenti controlli e produrre le risposte indicate:

1. Se il nome utente è la stringa vuota (""), stampare "Errore: nome utente non fornito".
2. Altrimenti, se l'età è minore di 0 o maggiore di 120, stampare "Errore: età non valida".
3. Altrimenti, se il livello di accesso non è uno tra 'guest', 'user' o 'admin', stampare "Errore: livello di accesso non riconosciuto".
4. Se tutti i dati sono corretti, valutare poi:
   - Se l'età è inferiore a 18 e il livello è 'admin', stampare "Accesso negato: minori non autorizzati come admin".
   - Se il livello è 'guest', stampare "Accesso limitato: utente guest" (indipendentemente dall'età).
   - Se il livello è 'user' e l'età è almeno 18, stampare "Accesso consentito: utente".
   - Se il livello è 'admin' e l'età è almeno 18, stampare "Accesso consentito: amministratore".

Questo esercizio richiede di gestire più condizioni in cascata verificando restrizioni sui dati di input e di usare correttamente elif per evitare ridondanze, garantendo un flusso chiaro e leggibile. Assicurati di usare un’adeguata indentazione e di evitare condizioni inutilmente complesse o ripetute.

Utilizza input() per acquisire i dati e le opportune conversioni di tipo per l'età. Stampa la risposta sulla console utilizzando print().

###**Esempi di Input/Output**
input:
"Alice"
25
admin <br>
output:
Accesso consentito: amministratore

input:
""
30
user <br>
output:
Errore: nome utente non fornito

input:
"Marco"
-5
user <br>
output:
Errore: età non valida

input:
"Luca"
19
superuser <br>
output:
Errore: livello di accesso non riconosciuto

input:
"Francesca"
16
admin <br>
output:
Accesso negato: minori non autorizzati come admin

input:
"Giorgio"
40
guest <br>
output:
Accesso limitato: utente guest

input:
"Sara"
18
user <br>
output:
Accesso consentito: utente

```python
# STEP 1: Richiesta nome utente, con controllo di validità
while True: # Il ciclo continua finché il nome inserito è valido
  user_name_str = input("Inserisci il tuo nome utente: ") # Richiesta del nome utente
  if len(user_name_str) == 0: # Controllo che la stringa non sia vuota
    print("Errore: nome utente non fornito") # Messaggio d'errore
  else:
    break # Se il nome è valido si passa allo step successivo

# STEP 2: Richiesta dell'età, con controllo di validità
while True: # Il ciclo continua finché il numero inserito è valido
  try: # Controllo che l'input sia un numero intero
    user_age_int = int(input("Inserisci la tua età: ")) # Casting dell'input a intero
    if user_age_int < 0 or user_age_int > 120: # Controllo che l'età sia valida
      print("Errore: età non valida") # Messaggio d'errore
    else:
      break # Se l'età è valida si passa allo step successivo
  except ValueError:
    print("Errore: età non valida") # Messaggio d'errore

# STEP 3: Richiesta del livello d'accesso (guest, user, admin), con controllo di validità
while True: # Il ciclo continua finché il livello inserito è valido
  user_access_level_str = input("Inserisci il tuo livello di accesso (guest, user, admin): ") # Richiesta del livello di accesso
  if user_access_level_str not in ["guest", "user", "admin"]: # Controllo che il livello sia valido
    print("Errore: livello di accesso non riconosciuto") # Messaggio d'errore
  else:
    break # Se il livello è valido si passa allo step successivo

# STEP 4: Se tutti i dati sono corretti, si valutano le seguenti condizioni:
# Se l'età è inferiore a 18 e il livello è 'admin', stampare "Accesso negato: minori non autorizzati come admin".
if user_age_int < 18 and user_access_level_str == "admin":
  print("Accesso negato: minori non autorizzati come admin")
# Se il livello è 'guest', stampare "Accesso limitato: utente guest" (indipendentemente dall'età).
elif user_access_level_str == "guest":
  print("Accesso limitato: utente guest")
# Se il livello è 'user' e l'età è almeno 18, stampare "Accesso consentito: utente".
elif user_access_level_str == "user" and user_age_int >= 18:
  print("Accesso consentito: utente")
# Se il livello è 'admin' e l'età è almeno 18, stampare "Accesso consentito: amministratore".
elif user_access_level_str == "admin" and user_age_int >= 18:
  print("Accesso consentito: amministratore")
```
