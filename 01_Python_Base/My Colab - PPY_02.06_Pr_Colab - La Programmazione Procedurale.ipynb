#**01-Principiante**
Scrivi un programma Python che definisca e utilizzi alcune funzioni semplici. In particolare:

1. Definisci una funzione chiamata `say_hello` che prende come parametro il nome di una persona e stampa un saluto nel formato "Ciao, <nome>!".

2. Definisci una funzione chiamata `calcola_area_triangolo` che riceve due parametri: base e altezza (numeri), e ritorna l'area del triangolo calcolata con la formula (base * altezza) / 2.

3. Nel programma principale, chiedi all'utente di inserire il proprio nome, poi la base e altezza di un triangolo.

4. Usa la funzione `say_hello` per salutare l'utente con il suo nome.

5. Usa la funzione `calcola_area_triangolo` per calcolare l'area con i valori inseriti.

6. Infine stampa il risultato dell'area formattandolo con due cifre decimali.

Ricorda che per leggere i dati da tastiera devi usare `input()`, e che i dati numerici letti da tastiera devono essere convertiti con il casting a `float` o `int` per poterli usare nei calcoli.

###**Esempi di Input/Output**
input:
Mario
10
5 <br>

L'area del triangolo con base 10.0 e altezza 5.0 è 25.00.

input:
Anna
7.5
3.2 <br>

L'area del triangolo con base 7.5 e altezza 3.2 è 12.00.

```python
# 1. Definizione della funzione 'say_hello' che prende come paramtero un nome
def say_hello(name):
  print(f"Ciao, {name}!")

# 2. Definizione della funzione 'calcola_area_triangolo' che riceve due
# parametri: base e altezza (numeri decimali), e ritorna l'area del triangolo
def calcola_area_triangolo(base, altezza):
  area = (base * altezza) / 2
  return area # Ritorna il valore calcolato

# 3. Definizione della funzione che verifica i numeri inseriti in input
def check_number(number):
  # Controlla che l'input sia un numero decimale positivo diverso da 0
  while True:
    try:
      number = float(number) # Casting dell'input in decimale

      if number == 0: # Gestione del valore 0
        print("Numero non valido: il numero inserito è 0")
        number = input("Inserisci un numero valido: ")
      elif number < 0: # Gestione dei valori negativi
        print("Numero non valido: il numero inserito è negativo")
        number = input("Inserisci un numero valido: ")
      else: # Se il numero è un decimale positivo, allora restituisce il valore
        return number
    except ValueError: # Gestione dell'errore del casting a decimale
      print("Il valore inserito non è un numero")
      number = input("Inserisci un numero valido: ")


# 4. Richiesta degli input
name = input("Inserisci il tuo nome: ") # Input per il nome (stringa)
base = check_number(input("Inserisci la base del triangolo: ")) # Input base validato
altezza = check_number(input("Inserisci l'altezza del triangolo: ")) # Input altezza validato


# 5. Invoca le funzioni e stampa l'output
say_hello(name) # Stampa il saluto
area = calcola_area_triangolo(base, altezza) # Calcola l'area

# Usa f-string con formattazione a 2 decimali per l'output
print(f"L'area del triangolo con base {base} e altezza {altezza} è {area:.2f}.")

```

```python
#==============================================================================
# DEFINIZIONE DELLE FUNZIONI
#==============================================================================

# 1. Funzione per il casting a float della lista numerica in input
def check_list(number_list):
  number_list = number_list.split() # Split della stringa in input
  float_list = [] # Inizializzazione della lista numerica

  for val in number_list: # Scorre la lista in input
    try:
      val = float(val) # Casting dell'input in decimale
      float_list.append(val) # Aggiunge il valore alla lista numerica
    except ValueError: # Gestione dell'errore del casting a decimale
      print(f"Il valore '{val}' inserito non è un numero")
      return False # Ritorna False
  return float_list # Se il casting a decimale è corretto, ritorna la lista numerica

# 2. Funzione 'calcola_media(lista)' della lista dei decimali precedentemente convertita
def calcola_media(float_list):
  mean = sum(float_list) / len(float_list)
  return mean # Ritorna la media

# 3. Funzione 'valore_massimo(lista) della lista dei decimali precedentemente convertita
def valore_massimo(float_list):
  max_num = float_list[0] # Inizializza il massimo con il primo elemento
  for num in float_list: # Scorre la lista in input
    if num > max_num: # Confronta il valore attuale con il massimo attuale
      max_num = num # Aggiorna in massimo attuale
  return max_num # Ritorna il massimo assoluto

# 4. Funzione 'stampa_lista_con_indice(lista)' della lista dei decimali precedentemente convertita
def stampa_lista_con_indice(float_list):
  for i, num in enumerate(float_list): # 'enumarate' crea la tupla (i, num)
    print(f"Elemento {i+1}: {num}") # Stampa della lista


#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

number_list = input("Inserisci una serie di numeri separati da spazi: ") # Richiesta della lista in input
float_list = check_list(number_list) # Casting della lista numerica

if float_list is not False: # Se il casting a decimale è corretta, il programma prosegue

  mean = calcola_media(float_list) # Calcolo della media
  print(f"\nMedia: {mean:.2f}\n") # Stampa della media

  max_num = valore_massimo(float_list) # Calcolo del massimo
  print(f"Massimo: {max_num:.2f}\n") # Stampa del massimo

  stampa_lista_con_indice(float_list) # Stampa della lista

```

#**04-Difficile**
In questa esercitazione realizzerai un programma modulare che gestisce dati numerici inseriti dall'utente, per calcolare statistiche complesse usando funzioni definite dal programmatore. Il programma dovrà eseguire le seguenti operazioni:

1. Acquisire da tastiera una lista di numeri interi o float, terminando l’inserimento con una parola chiave (esempio: 'fine').

2. Implementare almeno tre funzioni: una per calcolare la media dei numeri, una per trovare il massimo, una per trovare il minimo. Le funzioni dovranno ricevere come parametro la lista di numeri e restituire rispettivamente il valore calcolato.

3. Utilizzare il ciclo while o for per gestire l'inserimento e successivamente iterare per i calcoli richiesti.

4. Gestire eventuali casi particolari come lista vuota (nessun numero inserito) stampando un messaggio informativo.

5. Stampare i risultati formattati con precisione fino a due cifre decimali utilizzando f-string.

6. Separare chiaramente la logica di input, elaborazione (funzioni), e output nel codice, richiamando le funzioni definite.

Questa esercitazione ti aiuterà a ragionare in modo top-down e a praticare la divisione del codice in funzioni con responsabilità singole, migliorando la riutilizzabilità e la leggibilità del codice applicando i concetti della programmazione procedurale.

###**Esempi di Input/Output**
input:
10
5.5
3
fine <br>
output:
Media: 6.17
Massimo: 10.00
Minimo: 3.00

input:
fine <br>
output:
Nessun numero inserito, impossibile calcolare le statistiche.

input:
-2
-8
-0.5
fine <br>
output:
Media: -3.50
Massimo: -0.50
Minimo: -8.00

```python
#==============================================================================
# DEFINIZIONE DELLE FUNZIONI
#==============================================================================

# 1. Funzione 'check_lit' per verificare che l'input sia float.
# a. Si realizza il casting a float di ogni elemento della lista in input.
# b. Si gestisce il caso di input vuoto (nessun numero inserito).
def check_list(number_list):
  float_list = [] # Inizializzazione della lista numerica
  # Controlla se la lista è vuota o se il primo elemento è la parola 'fine'
  if len(number_list) == 0 or number_list[0].lower() == "fine":
    print("Nessun numero inserito, impossibile calcolare le statistiche.")
    return None # Ritorna None
  # Gestione della parola chiave 'fine' solo in ultima posizione (case insensitive)
  if number_list[-1].lower() == "fine": # 'lower' rende 'fine' 'case insensitive'
    number_list.pop(-1) # Elimina 'fine' solo dall'ultima posizione
  # Se l'input non è vuoto si procede con il casting a float
  for val in number_list: # Scorre la lista in input
    try:
      val = float(val) # Casting dell'input a float
      float_list.append(val) # Aggiunge il valore alla lista numerica
    except ValueError: # Gestione dell'errore del casting a float
      print(f"Il valore '{val}' inserito non è un numero")
      return None # Ritorna None
  # Se il casting a float è corretto, ritorna la lista numerica
  return float_list

# 2. Funzione 'calcola_media' della lista di float
# Si utilizza il ciclo 'for' come richiesto dall'esercizio
def calcola_media(float_list):
  sum_num = 0 # Inizializza la somma
  count = 0 # Inizializzazione del contatore
  for num in float_list: # Scorre la lista in input
    sum_num += num # Aggiunge il valore alla somma
    count += 1 # Incrementa il contatore
  # Non esiste la possibilità di divisione per 0 in quanto la funzione
  # 'check_list' non permette il casting di liste vuote
  mean = sum_num / count # Calcola la media
  return mean # Ritorna la media

# 3. Funzione 'max_num' della lista di float.
# Si utilizza il ciclo 'for' come richiesto dall'esercizio.
def max_num(float_list):
  is_max = float_list[0] # Inizializza il massimo con il primo elemento
  for num in float_list: # Scorre la lista in input
    if num > is_max: # Confronta il valore attuale con il massimo attuale
      is_max = num # Aggiorna in massimo attuale
  return is_max # Ritorna il massimo

# 4. Funzione 'min_num' della lista di float.
# Si utilizza il ciclo 'while' come alternativa al ciclo 'for'.
def min_num(float_list):
  is_min = float_list[0] # Inizializza il minimo con il primo elemento
  i = 0
  while i < len(float_list): # Scorre la lista in input
    if float_list[i] < is_min: # Confronta il valore attuale con il minimo attuale
      is_min = float_list[i] # Aggiorna in minimo attuale
    i += 1 # Incrementa l'indice
  return is_min # Ritorna il minimo


#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

# 1. Richiesta dell'input.
# La lista di input è composta de una serie di valori separati da spazi e
# termina con la parola 'fine'.
number_list = input("Inserisci una serie di numeri separati da spazi "
                    "(fine per terminare): ").split()

# 2. Invoca 'check_list' per l'analisi dell'input e il casting a float.
float_list = check_list(number_list)

# 3. Solo se il casting è corretto si procede con il calcolo delle statistiche.
if float_list is not None:
  mean_value = calcola_media(float_list) # Calcolo della media
  max_value = max_num(float_list) # Calcolo del massimo
  min_value = min_num(float_list) # Calcolo del minimo

  # Stampa delle statistiche: ognuna su una nuova riga
  print(f"\nMedia: {mean_value:.2f}\nMassimo: {max_value:.2f}\nMinimo: {min_value:.2f}")
```

#**05-Esperto**
Scrivi un programma Python che permetta di gestire una lista di studenti con i loro voti finali di un esame, implementando la logica utilizzando funzioni secondo il paradigma della programmazione procedurale. Il programma dovrà svolgere le seguenti operazioni:

1. Definisci una funzione per leggere da tastiera un numero intero positivo N, che rappresenta il numero di studenti da inserire. Usa un ciclo per garantire che il valore sia valido (maggiore di zero).

2. Definisci una funzione che, dato N, accetta in input da tastiera N coppie di dati: nome dello studente (stringa) e voto finale (numero intero da 0 a 30). Memorizza queste coppie come tuple dentro una lista.

3. Definisci una funzione che, data la lista degli studenti, calcoli e ritorni la media aritmetica dei voti.

4. Definisci una funzione che, data la lista, restituisca il nome dello studente con il voto più alto. Nel caso di parità, restituisci il primo incontrato.

5. Definisci una funzione che stampi tutti gli studenti con il relativo voto, in ordine di inserimento, formattando la stampa con f-string per allineare i dati.

6. Nel programma principale, usa le funzioni definite per:
  - Ottenere N.
  - Ottenere i dati degli studenti.
  - Calcolare e stampare la media dei voti.
  - Stampare il nome e il voto dello studente più bravo.
  - Stampare l'elenco completo degli studenti e voti.

Assicurati la corretta gestione di eventuali input errati e che ogni funzione abbia una singola responsabilità ben definita. Cerca di usare argomenti con valori di default e keyword arguments laddove opportuno, e sfrutta i valori di ritorno delle funzioni per propagare le informazioni.

Non usare altri paradigmi di programmazione, classi o eccezioni ancora non affrontate nel corso.

###**Esempi di Input/Output**
input:
3
Alice
28
Bob
30
Carla
27 <br>
output:
Media dei voti: 28.33
Studente con il voto più alto: Bob (30)
Elenco studenti:
1) Alice - 28
2) Bob - 30
3) Carla - 27

input:
2
Luca
15
Maria
15 <br>
output:
Media dei voti: 15.0
Studente con il voto più alto: Luca (15)
Elenco studenti:
1) Luca - 15
2) Maria - 15

```python
#==============================================================================
# DEFINIZIONE DELLE FUNZIONI
#==============================================================================

# 1. Funzione 'check_student_qty' per leggere da tastiera un numero intero
# positivo N, che rappresenta il numero di studenti da inserire.
def check_student_qty():
  while True:
    try:
      student_qty = int(input("Inserisci il numero di studenti: "))
      if student_qty > 0:
        return student_qty
      else:
        print("Il numero di studenti deve essere maggiore di 0")
    except ValueError:
      print("Il valore inserito non è un numero intero")

# 2. Funzione 'get_student' che, dato N, accetta in input da tastiera N coppie
# di dati: nome dello studente (stringa) e voto finale (numero intero da 0 a 30).
# Memorizza queste coppie come tuple dentro una lista.
def get_student(student_qty):
  student_list = [] # Inizializzazione della lista di studenti
  for i in range(student_qty): # Scorre la lista in input
    # Richiesta del nome
    name = input(f"Inserisci il nome dello studente {i+1}: ")
    # Richiesta del voto
    while True: # Il ciclo si interrompe solo a fronte di un evento
      try:
        grade = int(input(f"Inserisci il voto dello studente {i+1}: "))
        if 0 <= grade <= 30:
          student_list.append((name, grade)) # Aggiunge il valore alla lista
          break # Interrompe ed esce dal ciclo 'while'
        else:
          print("Il voto deve essere compreso tra 0 e 30") # Il ciclo continua
      except ValueError:
        print("Il valore inserito non è un numero intero") # Il ciclo continua
  return student_list # Ritorna la lista di studenti

# 3. Funzione 'mean_grades' della lista di float.
def mean_grades(student_list):
  sum_grade = 0 # Inizializza la somma
  for student in student_list: # Scorre la lista in input
    sum_grade += student[1] # Aggiunge il valore alla somma
  mean_grade = sum_grade / len(student_list) # Calcola
  return mean_grade # Ritorna la media

# 4. Funzione 'better_student' restituisce il nome dello studente con il voto
# più alto. Nel caso di parità, restituisce il primo incontrato.
def better_student(student_list):
  best_student = student_list[0] # Inizializza il massimo con il primo elemento
  for student in student_list: # Scorre la lista in input
    if student[1] > best_student[1]: # Confronta il valore attuale con il massimo attuale
      best_student = student # Aggiorna in massimo
  return best_student # Ritorna il massimo

# 5. Funzione 'student_print' stampa tutti gli studenti con il relativo voto,
# in ordine di inserimento, usando enumerate per l'indice.
def student_print(student_list):
  for i, student in enumerate(student_list): # 'enumarate' crea la tupla (i, student)
    print(f"{i+1:2}) {student[0]:8} - {student[1]:4}") # Stampa della lista


#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

# Invoca le funzioni
student_qty = check_student_qty() # Ottenimento del numero di studenti
student_list = get_student(student_qty) # Ottenimento della lista di studenti
mean_grade = mean_grades(student_list) # Calcolo della media
best_student = better_student(student_list) # Ottenimento dello studente con il voto più alto

# Stampa dei risultati
print(f"\nMedia dei voti: {mean_grade:.2f}") # Stampa della media
print(f"Studente con il voto più alto: {best_student[0]} ({best_student[1]})") # Stampa dello studente con voto maggiore
student_print(student_list) # Stampa deli studenti in ordine di inserimento
```
