In questa esercitazione creerai una classe Python chiamata Triangolo che rappresenta un triangolo con tre lati e un'altezza. Dovrai definire il costruttore __init__ per ricevere i valori dei tre lati e dell'altezza e salvarli come attributi d'istanza. Implementa due metodi: perimetro(), che calcola e ritorna la somma dei tre lati, e area(), che calcola e ritorna l'area del triangolo usando la formula base * altezza / 2 (prendendo come base uno dei lati). Infine, crea un'istanza della classe usando valori a tua scelta, e stampa sia il perimetro sia l'area richiamando i rispettivi metodi sull'oggetto creato. Ricorda di usare self per accedere agli attributi nell'ambito della classe. Usa la funzione print per mostrare i risultati su schermo.

###**Esempi di Input/Output**
input:
(lato1=3, lato2=4, lato3=5, altezza=4) <br>
output:
Perimetro: 12
Area: 6.0

input:
(lato1=6, lato2=6, lato3=6, altezza=5.2)<br>
output:
Perimetro: 18
Area: 15.6

```python
#==============================================================================
# DEFINIZIONE DELLE FUNZIONI E DELLE CLASSI
#==============================================================================

# Funzione per casting a float degli input:
# - I valori vengono controllati uno per uno per ordine d'inserimento
# - La funzione itera l'input fino all'inserimento corretto di tutti i dati
def chec_float(val):

  """
  Funzione per controllare il casting a float degli input
  """

  # Ciclo 'while' per inserimento corretto dell'input.
  # Il ciclo continua finché non viene inserito un numero valido, positivo e
  # diverso da zero
  while True:
    # Gestione del valore assente
    if val == "":
      val = input("Valore assente. Inserire un numero: ")
    else:
      # Se il valore è presente, effettua il casting a float, con gestione
      # dell'errore
      try:
        val = float(val)
        # Gestione del valore nullo o negativo
        if val == 0 or val < 0:
         val = input ("Inserire un numero positivo diverso da zero: ")
        # Se il valore è diverso da zero e positivo, la funzione restituisce
        # il valore convertito a float
        else:
          return val
          break
      # Gestione dell'input non numerico
      except ValueError:
        val = input(f"'{val}' non è un numero. Inserisci un numero: ")


# Definizione della classe padre 'Triangolo'
class Triangolo():

  """
  Classe che rappresenta un triangolo con tre lati e un'altezza.
  """

  #Definizione del costruttore
  def __init__(self, lato1, lato2, lato3, altezza):

    """
    Costruttore della classe Triangolo.

    Args:
      lato1 (float): Lunghezza del primo lato.
      lato2 (float): Lunghezza del secondo lato.
      lato3 (float): Lunghezza del terzo lato.
      altezza (float): Altezza del triangolo.
    """

    self.lato1 = lato1
    self.lato2 = lato2
    self.lato3 = lato3
    self.altezza = altezza

  # Definizione del metodo 'perimetro'
  def perimetro(self):

    """
    Calcola il perimetro del triangolo.

    Returns:
      float: La somma dei tre lati del triangolo.
    """

    return self.lato1 + self.lato2 + self.lato3

  # Definizione del metodo 'area'
  def area(self):

    """
    Calcola l'area del triangolo.

    Returns:
      float: L'area del triangolo.
    """

    return (self.lato1 * self.altezza) / 2

  # Definizione della funzione print
  def __repr__(self):

    """
    Ritorna i valori di perimetro ed area del triangolo.

    Returns:
      str: Perimetro ed area del tirangolo.
    """

    info = f"\nDato il triangolo di lati {self.lato1}, {self.lato2}, {self.lato3} e altezza {self.altezza}\n"
    info +=f"Il perimetro è {self.perimetro():.2f}\n"
    info += f"L'area è {self.area():.2f}"

    return info

#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

# Richiesta dei valori di input
lato1 = chec_float(input("Inserisci il primo lato: "))
lato2 = chec_float(input("Inserisci il secondo lato: "))
lato3 = chec_float(input("Inserisci il terzo lato: "))
altezza = chec_float(input("Inserisci l'altezza: "))

# Istanziamento della classe 'Triangolo'
triangolo = Triangolo(lato1, lato2, lato3, altezza)

# Stampa dei risultati
print(triangolo)
```

```python
#==============================================================================
# DEFINIZIONE DELLE COSTANTI CONDIVISE
#==============================================================================

# Definizione dei valori di default
DEFAULT_LATO1 = 2
DEFAULT_LATO2 = 3
DEFAULT_LATO3 = 4
DEFAULT_ALTEZZA = 5


#==============================================================================
# DEFINIZIONE DELLE FUNZIONI E DELLE CLASSI
#==============================================================================

# Funzione per casting a float degli input.
def check_float(value, default):

  """
  Funzione per la conversione dell'input a float

  Restituisce il valore di default se l'input è vuoto, altrimenti
  esegue il casting a float
  """

  # Gestione del valore assente (non inserito):
  # Se l'utente non digita alcun valore, allora si accettano i valori di default
  if value == "":
    return default

  # Se l'input non è vuoto, allora esegue il casting a float
  value = float(value)

  # Gestione dell'input zero e negativo
  if value <= 0:
    raise ValueError("Il numero deve essere maggiore di zero")

  return value # Se la conversione è corretta resituisce il valore


# Definizione della funzione per ricevere l'input
def get_float(prompt, default):
  """
  Funzione per la ricezione dell'input da tastiera

  Ripete la richiesta fino a che l'input non è valido
  """

  while True:
    # Chiede l'inserimento del valore di input
    value = input(prompt)
    try:
      # Esegue la conversione a float chiamando la funzione 'check_float'
      value = check_float(value, default)
      return value # Se la conversione è corretta, restituisce il valore
    except ValueError:
      # Se la conversione fallisce, stampa l'errore
      print("Valore non valido. Inserisci un numero maggiore di zero: ")



# Definizione della classe padre 'Triangolo'
class Triangolo():
  """
  Classe che rappresenta un triangolo con tre lati e un'altezza.
  """

  #Definizione del costruttore
  def __init__(
      self,
      lato1=DEFAULT_LATO1,
      lato2=DEFAULT_LATO2,
      lato3=DEFAULT_LATO3,
      altezza=DEFAULT_ALTEZZA
      ):

    """
    Costruttore della classe Triangolo.

    Args:
      lato1 (float): Lunghezza del primo lato.
      lato2 (float): Lunghezza del secondo lato.
      lato3 (float): Lunghezza del terzo lato.
      altezza (float): Altezza del triangolo.
    """
    # Definizione degli attributi
    self.lato1 = lato1
    self.lato2 = lato2
    self.lato3 = lato3
    self.altezza = altezza

  # Definizione del metodo per il calcolo del perimetro
  def perimetro(self):

    """
    Calcola il perimetro del triangolo.

    Returns:
      float: La somma dei tre lati del triangolo.
    """

    return self.lato1 + self.lato2 + self.lato3

  # Definizione del metodo per il calcolo dell'area
  def area(self):

    """
    Calcola l'area del triangolo.

    Returns:
      float: L'area del triangolo.
    """

    return (self.lato1 * self.altezza) / 2

  # Definizione della funzione print
  def __repr__(self):

    """
    Ritorna i valori di perimetro ed area del triangolo.

    Returns:
      str: Perimetro ed area del tirangolo.
    """

    info = (
        f"\nDato il triangolo di lati {self.lato1}, {self.lato2}, {self.lato3}"
        f" e altezza {self.altezza}\n"
    )
    info +=f"Il perimetro è {self.perimetro():.2f}\n"
    info += f"L'area è {self.area():.2f}"

    return info

#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

# Definizione dei valori di default
DEFAULT_LATO1 = 2
DEFAULT_LATO2 = 3
DEFAULT_LATO3 = 4
DEFAULT_ALTEZZA = 5

# Richiesta dei valori di input:
# 'get_float' riceve 2 argomenti 'prompt' e il valore di default
l1 = get_float("Inserisci il primo lato: ", DEFAULT_LATO1)
l2 = get_float("Inserisci il secondo lato: ", DEFAULT_LATO2)
l3 = get_float("Inserisci il terzo lato: ", DEFAULT_LATO3)
h = get_float("Inserisci l'altezza: ", DEFAULT_ALTEZZA)

# Istanziamento della classe 'Triangolo'
triangolo = Triangolo(l1, l2, l3, h)

# Stampa dei risultati
print(triangolo)
```

```python
#==============================================================================
# 1. DEFINIZIONE DELLE FUNZIONI E DELLE CLASSI
#==============================================================================


# -----------------------------------------------------------------------------
# 1.1 SUPERCLASSE 'Shape'
# -----------------------------------------------------------------------------

class Shape ():
  """
  Classe base per rappresentare forme geometriche.
  """

  def __init__(self, lati, altezza = None):
    """
    Costruttore della classe Shape.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati della forma.
      altezza (float): Il valore dell'altezza di un triangolo. Il valore
      di default è None.
    """
    self._lati = lati
    self._altezza = altezza

  def perimetro(self):
    """
    Calcola il perimetro della forma.

    Questo metodo verrà sovrascritto dalle classi figlie
    """
    pass

  def area(self):
    """
    Calcola l'area della forma.

    Questo metodo verrà sovrascritto dalle classi figlie
    """
    pass

  def __repr__(self):
    """
    Ritorna una rappresentazione testuale della forma.

    Questo metodo verrà sovrascritto dalle classi figlie che aggiungeranno
    maggiori dettagli delle figure geometriche specifiche.
    """
    info_shape = (
        f"<Shape {self.__class__.__name__}: lati={self._lati}, "
        f"altezza={self._altezza}>"
    )
    return info_shape


# -----------------------------------------------------------------------------
# 1.2 CLASSE 'Tiangle' - Sottoclasse di 'Shape'
# -----------------------------------------------------------------------------

class Triangle(Shape):
  """
  Classe per rappresentare un triangolo.
  """

  def __init__(self, lati, altezza):
    """
    Costruttore della classe Triangle.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del triangolo.
      altezza (float): Il valore dell'altezza del triangolo.
    """
    super().__init__(lati, altezza)

  def perimetro(self):
    """
    Calcola il perimetro del triangolo.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del triangolo.
      altezza (float): Il valore dell'altezza del triangolo.

    Returns:
      float: Il perimetro del triangolo.
    """
    perimetro = sum(self._lati)
    return perimetro

  def area(self):
    """
    Calcola l'area del triangolo.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del triangolo
      altezza (float): Il valore dell'altezza del triangolo.

    Returns:
      float: L'area del triangolo.
    """
    base = self._lati[0]
    area = (base * self._altezza) / 2
    return area


# -----------------------------------------------------------------------------
# 1.3 CLASSE 'Square' - Sottoclasse di 'Shape'
# -----------------------------------------------------------------------------

class Square(Shape):
  """
  Classe per rappresentare un quadrato.
  """

  def __init__(self, lati):
    """
    Costruttore della classe Square.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del quadrato.
    """
    super().__init__(lati)

  def perimetro(self):
    """
    Calcola il perimetro del quadrato.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del quadrato.

    Returns:
      float: Il perimetro del quadrato.
    """
    perimetro = 4 * self._lati[0]
    return perimetro

  def area(self):
    """
    Calcola l'area del quadrato.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del quadrato.

    Returns:
      float: L'area del quadrato.
    """
    area = self._lati[0] ** 2
    return area


# -----------------------------------------------------------------------------
# 1.4 CLASSE 'Rectangle' - Sottoclasse di 'Shape'
# -----------------------------------------------------------------------------

class Rectangle(Shape):
  """
  Classe per rappresentare un rettangolo.
  """

  def __init__(self, lati):
    """
    Costruttore della classe Rectangle.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del rettangolo.
    """
    super().__init__(lati)

  def perimetro(self):
    """
    Calcola il perimetro del rettangolo.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del rettangolo.

    Returns:
      float: Il perimetro del rettangolo.
    """
    perimetro = 2 * (self._lati[0] + self._lati[1])
    return perimetro

  def area(self):
    """
    Calcola l'area del rettangolo.

    Args:
      lati (tuple): Una tupla di numeri che rappresentano i lati del rettangolo.

    Returns:
      float: L'area del rettangolo.
    """
    area = self._lati[0] * self._lati[1]
    return area

#==============================================================================
# 2. CODICE PRINCIPALE
#==============================================================================

# 2.1 Istanziamento delle figure geometriche
triangolo = Triangle((3,4,5), 4)
quadrato = Square((5,))
rettangolo = Rectangle((6,4))

# 2.2 Stampa dei risutlati
print(triangolo)
print(f"Triangle perimeter: {triangolo.perimetro():.2f}")
print(f"Triangle area: {triangolo.area():.2f}")

print(f"\n{quadrato}")
print(f"Square perimeter: {quadrato.perimetro():.2f}")
print(f"Square area: {quadrato.area():.2f}")

print(f"\n{rettangolo}")
print(f"Rectangle perimeter: {rettangolo.perimetro():.2f}")
print(f"Rectangle area: {rettangolo.area():.2f}")
```
