# **01-Principiante**

Scrivi un programma Python che:

1. Chiede all'utente di inserire due numeri interi sulla stessa riga separati da uno spazio.
2. Divide il primo numero per il secondo.
3. Stampa il risultato della divisione.

Se l'utente inserisce valori non numerici o non esattamente due numeri, il programma deve mostrare un messaggio "Errore: devi inserire esattamente due numeri interi separati da uno spazio.".

Se si verifica una divisione per zero, il programma deve mostrare il messaggio "Errore: divisione per zero non consentita.".

In ogni caso (sia in caso di errore sia di successo), il programma deve stampare alla fine "Operazione terminata.".

Utilizza il costrutto try-except-finally per gestire le eccezioni e garantire che il messaggio finale venga sempre mostrato.

Non usare nessun argomento o modulo non ancora affrontato nei moduli seguiti fino ad ora.
<br><br>
### **Esempi di Input/Output**
input: 10 2 <br>
output: 5.0
Operazione terminata.

input: 10 0 <br>
output: Errore: divisione per zero non consentita.
Operazione terminata.

input: 5 a <br>
output: Errore: devi inserire esattamente due numeri interi separati da uno spazio.
Operazione terminata.

input: 3 <br>
output: Errore: devi inserire esattamente due numeri interi separati da uno spazio.
Operazione terminata.

```python
try:

  # Chiede all'utente di inserire due numeri interi separati da uno spazio.
  # Si esegue il casting sulla stessa riga di codice
  int_list = list(
      map(
          int,
          input(
              "Inserisci due numeri interi separati da uno spazio: "
              ).split()
          )
  )

  # Controllo se l'input contiene esattamente due elementi
  # Uso assert per sollevare AssertionError in caso contrario
  assert len(int_list) == 2, (
      "Errore: devi inserire esattamente due numeri "
      "interi separati da uno spazio."
  )

  # Divide il primo numero della lista per il secondo.
  c= int_list[0]/int_list[1]

  # Stampa il risultato della divisione.
  print(f"{c:.2f}")

# Gestione dell' input non convertibile a 'int'
except ValueError:
  print(
      "Errore: devi inserire esattamente due numeri interi "
      "separati da uno spazio."
      )

# Gestione dell'input diverso da 2 valori
except AssertionError as e:
  print(e)

# Gestione della divisione per zero
except ZeroDivisionError:
  print("Errore: divisione per zero non consentita.")

# Cattura eventuali altre eccezioni non previste per evitare il crash
except Exception as e:
  print(f"Errore: {e}")

# Gestione dell'uscita del blocco 'try'
finally:
  print("Operazione terminata.")
```

# **04-Difficile**

Scrivi un programma Python che:

1. Legga dall'utente una singola riga di input contenente esattamente due valori separati da spazio.
2. Controlli che ci siano esattamente due valori; se non sono due, sollevi una ValueError con messaggio personalizzato.
3. Convertisca entrambi i valori in interi, gestendo eventuali errori di conversione.
4. Esegua la divisione del primo numero per il secondo, gestendo la divisione per zero tramite eccezione.
5. Stampi il risultato con due cifre decimali.
6. In ogni caso, stampi un messaggio di chiusura "Operazione terminata." (usando un blocco finally).

Estendi il programma per anticipare e gestire questi casi di errore, stampando messaggi chiari e specifici:
- Numero sbagliato di input
- Valori non convertibili in interi
- Divisione per zero
- Qualsiasi altra eccezione imprevista

Non terminare il programma con un errore non gestito. Utilizza la gestione delle eccezioni con try/except/finally per assicurare la robustezza del programma.

Non usare ancora la definizione di funzioni o moduli.
<br><br>
### **Esempi di Input/Output**
input:
10 2 <br>
output:
5.00
Operazione terminata.

input:
8 0 <br>
output:
Errore: divisione per zero non permessa.
Operazione terminata.

input:
7 <br>
output:
Errore: devono essere inseriti esattamente due valori.
Operazione terminata.

input:
9 a <br>
output:
Errore: entrambi i valori devono essere numeri interi.
Operazione terminata.

input:
10 2 3 <br>
output:
Errore: devono essere inseriti esattamente due valori.
Operazione terminata.

```python
#===============================================================================
# DEFINIZIONE DELLE FUNZIONI
#===============================================================================

def get_input ():
  """
  Funzione che chiede l'iserimento dei dati in input

  Returns
    input_lits: list[str], lista dei valori inseriti dall'utente
  """
  input_list = list(
      input(
          "Inserisci due numeri interi separati da uno spazio: "
          ).split() # Separa i valori della stringa in input
  )
  return input_list

def check_count_list(input_list):
  """
  Verifica che la lista contenga esattamente due elementi.

  Args:
    input_list: list[str]

  Raises:
    ValueError: se la lista non contiene esattamente due elementi.

  Returns:
    int: numero di elementi nella lista.
  """
  if len(input_list) != 2:
    raise ValueError("Inserire unicamente 2 numeri interi")
  else:
    return len(input_list)


def int_list(input_list):
  """
  Funzione che converte gli elementi della lista in interi

  Args:
    input_list: list[str]

  Returns:
    int_list: list[int], lista dei valori convertiti in interi

  Raises:
    ValueError: se un elemento non è convertibile a 'int'
  """
  # Realizza il casting a 'int' della lista in input
  int_list = list(map(int, input_list))
  return int_list


def divisione(int_list):
  """
  Esegue la divisione del primo numero della lista per il secondo.

  Args:
    int_list (list(int))

  Returns:
    division: float, risultato della divisione

  Raises:
    ZeroDivisionError: se il secondo numero è zero.
  """
  division = int_list[0]/int_list[1]
  return division

def print_result(division):
  """
  Funzione che stampa il risultato della divisione con due cifre decimali.
  """
  print(f"{division:.2f}")


#===============================================================================
# CODICE PRINCIPALE
#===============================================================================

try:
  # Acquisizione dei valori.
  valori = get_input()

  # Verifica della quantità di valori inseriti.
  check_count_list(valori)

  # Conversione degli elementi in interi.
  numeri = int_list(valori)

  # Calcolo della divisione.
  div = divisione(numeri)

  # Visualizzazione del risultato.
  print_result(div)

except ValueError as e:
  print(e)

except ZeroDivisionError:
  print("Errore: divisione per zero non permessa.")

except Exception as e:
  print(f"Errore: {e}")

finally:
  print("Operazione terminata.")
```
