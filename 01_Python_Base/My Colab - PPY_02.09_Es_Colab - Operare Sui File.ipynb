# **1. Filtra i proverbi**
Leggi il contenuto del file proverbi.txt e scrivi su di un nuovo file chiamato proverbi_filtrati.txt solo i proverbi che iniziano per vocale o che solo più brevi di 25 caratteri.

```python
# Il file 'proverbi.txt' è stato caricato manualmente nella librerira locale
# Apre il file in modalità lettura
with open('proverbi.txt', 'r', encoding='utf-8') as f:
  # Legge il file linea per linea creando la lista (str) 'proverbi'
  proverbi = f.readlines()
  print(type(proverbi)) # Stampa di conferma che 'proverbi' è una lista

# Opera temporalmente sul nuovo file, in modalità scrittura
with open('proverbi_filtrati.txt', 'w', encoding='utf-8') as f2:
  # Itera su ogni elemento della lista 'proverbi'
  for proverbio in proverbi:
    proverbio= proverbio.strip()

    # Controlla che la linea non sia vuota
    if not proverbio:
      continue

    # Per ogni elemento della lista controlla se inizia con una vocale o se la
    # sua lunghezza (senza spazi) è < 25
    if proverbio[0].lower() in 'aeiou' or len(proverbio) < 25:
      # Se le condizioni sono rispettate, scrive il nuovo elemento nel file
        f2.write(f"{proverbio}\n")

# Apre temporalmente il nuovo file, per leggerne il contenuto
with open('proverbi_filtrati.txt', 'r') as f:
  print(f.readlines())
```

# **2. Analisi di Magazzino**
Apri il file 'shirts.csv' e calcola le seguenti statistiche su di esso:
1. Numero totale di prodotti presenti
2. Valore totale del magazzino (somma del prezzo di tutti i prodotti)
3. Prezzo medio dei prodotti
4. Numero di prodotti per ogni colore
5. Numero di prodotti per ogni taglia
Salva le statistiche all'interno di un file json chiamato 'shirts_stats.json'

```python
# Il file 'shirts.cvs' è stato caricato manualmente nella librerira locale

# Essendo un file 'csv' è necessario importare la corrispondente libreria Python
import csv

# Opera temporalmente sul file in modalità lettura
with open('shirts.csv', 'r', encoding='utf-8') as f:
  record = csv.reader(f)
  print(next(record)) # Stampa la prima linea per vedere il nome delle colonne
  next(record) # Salta l'intestazione del file

  # Inizializzazione delle variabili
  num_records = 0 # Inizializza la variabile che conta gli elementi nel csv
  tot_price = 0 # Inizializza la variabile che somma i prezzi
  count_color = {} # Inizializza un dizionario vuoto per contare i colori
  count_size = {} # Inizializza un dizionario vuoto per contare le taglie

  for riga in record: # Itera su ciascun elemento del file 'csv'
    num_records += 1 # Incrementa il contatore degli elementi
    tot_price += float(riga[3]) # Aggiunge il prezzo alla somma totale

    color = riga[2] # Assegna il colore alla variabile 'color'
    size = riga[1] # Assegna la taglia alla variabile

    count_color[color] = count_color.get(color, 0) + 1 # Aggiorna il contatore del colore
    count_size[size] = count_size.get(size, 0) + 1 # Aggiorna il contatore della taglia


# Stampa delle variabili
print(f"Il numero di record è: {num_records}")
print(f"Il valore totale del magazzino è: {tot_price:.2f}")
print(f"Il prezzo medio dei prodotti è: {tot_price/num_records:.2f}")
print(f"I colori presenti nel magazzino sono: {count_color}")
print(f"Le taglie presenti nel magazzino sono: {count_size}")
```
