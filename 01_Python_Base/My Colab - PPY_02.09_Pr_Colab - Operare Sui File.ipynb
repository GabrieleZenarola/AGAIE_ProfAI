Usa la modalità di apertura "r" per leggere e "w" per scrivere.

```python
# Importa la libreria Google Colab per caricare i file
from google.colab import files

# Importa Path per gestire file e percorsi
from pathlib import Path

# Carica il file in input
uploaded = files.upload()

# Assegna il percorso del file caricato alla variabile 'input_file'
input_file = Path(next(iter(uploaded)))

# Ciclo 'try' - 'finally' per eliminare il file dal repository di Google Colab.
try:
  # Apre il file in modalità di lettura ('r') e con encoding utf-8
  with open(input_file, 'r', encoding = 'utf-8') as file_txt:
    file_lines = file_txt.readlines()  # Legge tutte le righe del file e le assegna alla variabile 'lines'
    num_lines = len(file_lines)  # Calcola il numero di righe presenti nel file

    print("\nContenuto del file di input:")
    for line in file_lines:  # Itera su ogni riga del file e la stampa
      print(line, end = '') # Rispetta gli 'a capo' del file di input
    print(f"\nNumero di righe: {num_lines}")  # Stampa il numero di righe nel file

  # Crea un nuovo file 'output.txt' con le linee del file di input in ordine inverso (dall'ultima alla prima)
  with open('output.txt', 'w', encoding = 'utf-8') as output_file:
    for line in reversed(file_lines):  # Itera su ogni riga del file in ordine inverso
      output_file.write(line.rstrip('\r\n') + '\n') # Elimina eventuali caratteri di fine riga e aggiunge un nuovo carattere '\n'

  print("\nContenuto scritto in 'output.txt':")
  with open('output.txt', 'r', encoding = 'utf-8') as output_file:
    for line in output_file:  # Legge ogni riga del file 'output.txt'
      print(line, end = '') # Rispetta i '\n' aggiunti alla creazinoe del file di output


finally:
    input_file.unlink() # Elimina il file caricato dall'ambiente Colab
    print(f"\nFile '{input_file.name}' eliminato da Colab.")
```

### **Esempi di Input/Output**

<br>

(Nessuno, la scrittura su file è interna) <br>

(prodotti_aggiuntivi.csv creato con almeno due righe aggiuntive come esempio) <br>
id,nome,quantità,prezzo <br>
4,Giacca,8,50.0 <br>
5,Calze,30,3.5 <br>

```python
#==============================================================================
# IMPORTAZIONE DELLE LIBRERIE E DEFINIZIONE DELLE COSTANTI
#==============================================================================

# Importa il modulo per leggere e scrivere file CSV
import csv

# Definizione dei nomi dei file di input e output
POEM_FILE = "poesia.txt"
TEXT_OUTPUT_FILE = "output.txt"
PRODUCTS_FILE = "prodotti.csv"
ADDITIONAL_PRODUCTS_FILE = "prodotti_aggiuntivi.csv"

# Definizione del numero di righe da copiare dal file originale
LINES_TO_COPY = 3

# Definizione dell'intestazione del CSV e dei nuovi prodotti da aggiungere
PRODUCT_COLUMNS = ["id", "nome", "quantità", "prezzo"]

NEW_PRODUCTS = [
    [4, "Giacca", 8, 50.0],
    [5, "Calze", 30, 3.5]
]


#==============================================================================
# DEFINIZIONE DELLE FUNZIONI
#==============================================================================

# 1. Funzione per leggere il file di testo e stamparne il contenuto
def read_poem(file_path):
    """
    Legge un file di testo e stampa ogni riga senza aggiungere righe vuote.

    Args:
        file_path (str): Percorso del file di testo da leggere.

    Returns:
        list[str]: Lista delle righe originali lette dal file.
    """

    # Apre il file in modalità lettura.
    # Il blocco 'with' ne gestisce automaticamente la chiusura.
    with open(file_path, "r", encoding="utf-8") as text_file:
        lines = text_file.readlines()  # Memorizza tutte le righe in una lista

    # Itera sulle righe lette dal file
    for line in lines:

        # Elimina soltanto i caratteri di fine riga già presenti.
        # print() aggiunge il proprio carattere di nuova linea,
        # evitando così di introdurre righe vuote aggiuntive.
        print(line.rstrip("\r\n"))

    return lines  # Restituisce le righe originali per le operazioni successive


# 2. Funzione per copiare le prime righe in un nuovo file di testo
def write_first_lines(lines, file_path, number_of_lines):
    """
    Scrive in un file il numero richiesto di righe iniziali.

    Args:
        lines (list[str]): Lista delle righe lette dal file originale.
        file_path (str): Percorso del file di testo da creare.
        number_of_lines (int): Numero non negativo di righe da copiare.
    """

    # La modalità 'w' crea il file oppure ne sovrascrive il contenuto
    with open(file_path, "w", encoding="utf-8") as output_file:

        # Lo slicing seleziona le prime righe richieste.
        # Se il file contiene meno righe, vengono selezionate quelle disponibili.
        for line in lines[:number_of_lines]:

            # Elimina i caratteri di fine riga esistenti e aggiunge un solo '\n'.
            # Gestisce anche l'ultima riga del file se priva di un 'a capo'.
            output_file.write(line.rstrip("\r\n") + "\n")


# 3. Funzione per leggere il file CSV e stampare i prodotti
def print_products(file_path):
    """
    Legge un inventario CSV e stampa nome, quantità e prezzo dei prodotti.

    Args:
        file_path (str): Percorso del file CSV contenente l'inventario.
    """

    # Apre il file in modalità lettura.
    # newline="" lascia al modulo csv la gestione dei caratteri di fine riga.
    with open(file_path, "r", encoding="utf-8", newline="") as csv_file:
        records = csv.reader(csv_file)

        # Salta l'intestazione una sola volta.
        # Il valore di default None evita un errore se il file è vuoto.
        next(records, None)

        # Itera sulle righe successive all'intestazione
        for row in records:

            # Salta le righe completamente vuote
            if not row:
                continue

            # Assegna i valori delle quattro colonne a variabili descrittive
            product_id, name, quantity, price = row

            # Converte il prezzo a float per visualizzarlo con due decimali
            print(
                f"Nome: {name}, "
                f"Quantità: {quantity}, "
                f"Prezzo: €{float(price):.2f}"
            )


# 4. Funzione per aggiungere nuovi prodotti a un file CSV
def append_products(file_path, columns, products):
    """
    Aggiunge prodotti a un CSV, scrivendo l'intestazione solo se il file è vuoto.

    Args:
        file_path (str): Percorso del file CSV di destinazione.
        columns (list[str]): Nomi delle colonne dell'intestazione.
        products (list[list]): Lista delle righe dei prodotti da aggiungere.
    """

    # La modalità 'a' aggiunge dati alla fine senza cancellare quelli esistenti.
    # Se il file non esiste, viene creato automaticamente.
    with open(file_path, "a", encoding="utf-8", newline="") as csv_file:

        # In modalità append, la posizione iniziale è alla fine del file.
        # Se tell() restituisce zero, il file è vuoto.
        file_is_empty = csv_file.tell() == 0

        # Crea l'oggetto per scrivere le righe in formato CSV
        writer = csv.writer(csv_file)

        # Scrive l'intestazione soltanto se il file è nuovo o vuoto
        if file_is_empty:
            writer.writerow(columns)

        # Itera sui nuovi prodotti e aggiunge una riga per ciascuno
        for product in products:
            writer.writerow(product)


#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

# STEP 1: Legge la poesia, ne stampa il contenuto e memorizza le righe originali
poem_lines = read_poem(POEM_FILE)

# STEP 2: Scrive le prime tre righe nel file di testo di output
write_first_lines(poem_lines, TEXT_OUTPUT_FILE, LINES_TO_COPY)

# STEP 3: Legge l'inventario e stampa i prodotti senza l'intestazione
print_products(PRODUCTS_FILE)

# STEP 4: Aggiunge i nuovi prodotti al file CSV di destinazione.
# Ogni nuova esecuzione aggiunge nuovamente gli stessi prodotti.
append_products(
    ADDITIONAL_PRODUCTS_FILE,
    PRODUCT_COLUMNS,
    NEW_PRODUCTS
)
```

```python
#==============================================================================
# IMPORTAZIONE DELLE LIBRERIE
#==============================================================================

# Importa il modulo csv per leggere e scrivere i file dei prodotti
import csv


#==============================================================================
# LETTURA E FILTRO DELLE PAROLE
#==============================================================================

# Inizializza una lista per conservare le parole con più di 5 caratteri
filtered_words = []

# Apre il file di testo in modalità lettura
# Il context manager 'with' chiude automaticamente il file al termine del blocco
with open("words.txt", "r", encoding="utf-8") as input_file:

  for line in input_file:

    # Elimina il carattere di nuova riga ed eventuali spazi iniziali e finali
    word = line.strip()

    # Conserva solo le parole con più di 5 caratteri
    if len(word) > 5:
      filtered_words.append(word)

# Il numero di elementi della lista corrisponde al numero di parole filtrate
word_count = len(filtered_words)

print(f"Numero di parole con più di 5 caratteri: {word_count}")


#==============================================================================
# SCRITTURA DELLE PAROLE FILTRATE
#==============================================================================

# Apre il file di destinazione in modalità scrittura
# Se il file esiste già, il suo contenuto viene sostituito
with open("filtered_words.txt", "w", encoding="utf-8") as output_file:

  for word in filtered_words:

    # Aggiunge una nuova riga per separare le parole nel file di destinazione
    output_file.write(word + "\n")


#==============================================================================
# LETTURA DEI PRODOTTI E CALCOLO DEL PREZZO MEDIO DEI PRODOTTI ROSSI
#==============================================================================

# Conserva tutti i prodotti per poterli confrontare successivamente con la media
products = []

# Inizializza la somma dei prezzi e il contatore dei prodotti rossi
red_total = 0
red_count = 0

# Apre il file CSV in modalità lettura
with open("products.csv", "r", encoding="utf-8", newline="") as input_file:

  # Ogni riga viene letta come un dizionario:
  # i nomi delle colonne diventano le chiavi
  reader = csv.DictReader(input_file)

  # Conserva l'intestazione per mantenere la stessa struttura nel file di output
  fieldnames = reader.fieldnames

  for product in reader:

    # Conserva il prodotto con i valori originali letti dal CSV
    products.append(product)

    # Normalizza il colore per riconoscere anche valori come "Rosso" o " rosso "
    if product["colore"].strip().lower() == "rosso":

      # I valori letti dal CSV sono stringhe:
      # converte il prezzo in float per poter eseguire il calcolo
      red_total += float(product["prezzo"])
      red_count += 1


#==============================================================================
# SCRITTURA DEI PRODOTTI CON PREZZO SUPERIORE ALLA MEDIA
#==============================================================================

# Verifica che esista almeno un prodotto rosso per evitare la divisione per zero
if red_count > 0:

  average_price = red_total / red_count

  # Mostra la media con quattro cifre decimali
  print(f"Prezzo medio prodotti rossi: {average_price:.4f}")

  # newline="" evita l'inserimento di righe vuote aggiuntive nel CSV
  with open("expensive_products.csv", "w",
            encoding="utf-8", newline="") as output_file:

    # Utilizza le stesse colonne e lo stesso ordine del file originale
    writer = csv.DictWriter(output_file, fieldnames=fieldnames)

    # Scrive l'intestazione del file CSV
    writer.writeheader()

    for product in products:

      # Confronta il prezzo di ogni prodotto con la media dei prodotti rossi
      if float(product["prezzo"]) > average_price:
        writer.writerow(product)

else:

  print("Non sono presenti prodotti rossi: impossibile calcolare la media.")
  print("Il file expensive_products.csv non viene creato.")
```

```python
#==============================================================================
# IMPORTAZIONE DELLE LIBRERIE
#==============================================================================

import csv
import json


#==============================================================================
# PARTE 1 - LETTURA, NORMALIZZAZIONE E SCRITTURA DEI NOMI
#==============================================================================

# Apre contemporaneamente il file originale in lettura e quello nuovo in scrittura
# Entrambi i file vengono chiusi automaticamente al termine del blocco 'with'
with open("file.txt", "r", encoding="utf-8") as input_file, \
     open("file_normalizzato.txt", "w", encoding="utf-8") as output_file:

  for line in input_file:

    # Elimina gli spazi iniziali/finali e il carattere di nuova riga
    # Converte in maiuscolo l'iniziale di ogni parola e in minuscolo le altre lettere
    name = line.strip().title()

    # Ignora eventuali righe vuote, che non rappresentano nomi
    if name != "":

      # Crea una stringa senza spazi per calcolare il numero di caratteri
      # Il nome normalizzato conserva invece gli eventuali spazi interni
      character_count = len(name.replace(" ", ""))

      # Scrive il nome e il numero di caratteri, quindi passa alla riga successiva
      output_file.write(f"{name} {character_count}\n")

print("File file_normalizzato.txt creato.")


#==============================================================================
# PARTE 2 - FILTRO DEI PRODOTTI CSV E CALCOLO DEL PREZZO CON IVA
#==============================================================================

# Definisce le categorie ammesse e l'aliquota IVA
allowed_categories = ["Abbigliamento", "Accessori"]
vat_rate = 0.22

# newline="" permette al modulo csv di gestire correttamente le nuove righe
with open("shoes.csv", "r", encoding="utf-8", newline="") as input_file, \
     open("shoes_filtrato.csv", "w",
          encoding="utf-8", newline="") as output_file:

  # Legge ogni prodotto come un dizionario usando l'intestazione come chiavi
  reader = csv.DictReader(input_file)

  # Crea l'intestazione di output mantenendo le colonne originali
  # e aggiungendo la nuova colonna alla fine
  fieldnames = reader.fieldnames + ["Prezzo IVA"]

  writer = csv.DictWriter(output_file, fieldnames=fieldnames)
  writer.writeheader()

  for product in reader:

    # Converte il prezzo da stringa a numero per eseguire confronti e calcoli
    price = float(product["prezzo"])

    # Elimina eventuali spazi iniziali/finali dalla categoria
    category = product["categoria"].strip()

    # Entrambe le condizioni devono essere soddisfatte
    if price > 20 and category in allowed_categories:

      price_with_vat = price * (1 + vat_rate)

      # Formatta il prezzo con IVA con due cifre decimali
      product["Prezzo IVA"] = f"{price_with_vat:.2f}"

      # Scrive il prodotto mantenendo anche tutti i suoi campi originali
      writer.writerow(product)

print("File shoes_filtrato.csv creato.")


#==============================================================================
# PARTE 3 - LETTURA, AGGIORNAMENTO E SCRITTURA DELL'INVENTARIO JSON
#==============================================================================

# Carica il contenuto JSON nelle corrispondenti strutture dati Python:
# un dizionario che contiene, nella chiave "prodotti", una lista di dizionari
with open("inventory.json", "r", encoding="utf-8") as input_file:
  inventory = json.load(input_file)

# Accede alla lista dei prodotti e aggiorna ogni dizionario
for product in inventory["prodotti"]:

  # I valori numerici del JSON vengono già caricati come numeri Python
  total_value = product["quantita"] * product["prezzo_unitario"]

  # Aggiunge il nuovo campo al dizionario del prodotto
  product["valore_totale"] = total_value

# Salva l'intero inventario aggiornato in un nuovo file
with open("inventory_aggiornato.json", "w", encoding="utf-8") as output_file:

  # indent=2 rende leggibile la struttura del file
  # ensure_ascii=False conserva direttamente eventuali caratteri accentati
  json.dump(inventory, output_file, indent=2, ensure_ascii=False)

print("File inventory_aggiornato.json creato.")
```

```python
#==============================================================================
# IMPORTAZIONE DELLE LIBRERIE
#==============================================================================

import csv
import json


#==============================================================================
# DEFINIZIONE DELLE FUNZIONI
#==============================================================================

#------------------------------------------------------------------------------
# PARTE 1 - LETTURA COMPLETA E LETTURA RIGA PER RIGA
#------------------------------------------------------------------------------

def read_document(input_path):
  """
  Legge e stampa un file di testo prima interamente, poi riga per riga.

  Args:
    input_path (str): Percorso del file di testo da leggere.
  """

  with open(input_path, "r", encoding="utf-8") as input_file:

    # read() legge tutto il contenuto e porta il cursore alla fine del file
    content = input_file.read()

    print("CONTENUTO COMPLETO:")
    print(content)

    # Riporta il cursore all'inizio per consentire una seconda lettura
    input_file.seek(0)

    print("LETTURA RIGA PER RIGA:")

    # readlines() restituisce una lista delle righe, conservando i newline
    for line in input_file.readlines():

      # Rimuove il newline finale prima che print() ne aggiunga uno
      print(line.rstrip("\n"))


#------------------------------------------------------------------------------
# PARTE 2 - SCRITTURA E AGGIORNAMENTO DEL FILE DI TESTO
#------------------------------------------------------------------------------

def write_and_append_text(output_path):
  """
  Scrive tre frasi, ne aggiunge una in append e stampa il file aggiornato.

  Args:
    output_path (str): Percorso del file di testo da creare o sovrascrivere.
  """

  phrases = [
    "Chi va piano va sano e va lontano.",
    "Sbagliando si impara.",
    "La pratica rende migliori."
  ]

  # La modalità "w" crea il file o sostituisce il contenuto esistente
  with open(output_path, "w", encoding="utf-8") as output_file:

    for phrase in phrases:
      output_file.write(phrase + "\n")

  # La modalità "a" aggiunge testo alla fine senza cancellare quello presente
  with open(output_path, "a", encoding="utf-8") as output_file:
    output_file.write("Ogni piccolo passo ti avvicina al tuo obiettivo.\n")

  with open(output_path, "r", encoding="utf-8") as input_file:
    print("CONTENUTO AGGIORNATO DI output.txt:")
    print(input_file.read())


#------------------------------------------------------------------------------
# PARTE 3 - LETTURA E SOVRASCRITTURA DEL FILE CSV
#------------------------------------------------------------------------------

def filter_shirts(csv_path):
  """
  Sovrascrive il CSV conservando l'intestazione e i prodotti con prezzo < 20.

  Args:
    csv_path (str): Percorso del file CSV da leggere e aggiornare.
  """

  rows = []
  filtered_rows = []

  with open(csv_path, "r", encoding="utf-8", newline="") as input_file:

    reader = csv.reader(input_file)

    # Legge e conserva l'intestazione, escludendola dalle righe dei prodotti
    header = next(reader)

    for row in reader:

      # Ignora le righe completamente vuote
      if row:

        # csv.reader restituisce liste: le converte in tuple come richiesto
        rows.append(tuple(row))

  # Completa i confronti prima di aprire il file in modalità scrittura
  # Se un prezzo non è convertibile, l'errore avviene prima della sovrascrittura
  for row in rows:
    price = float(row[3])

    if price < 20:
      filtered_rows.append(row)

  # Tutti i dati necessari sono già in memoria: ora può sovrascrivere il file
  with open(csv_path, "w", encoding="utf-8", newline="") as output_file:

    writer = csv.writer(output_file)
    writer.writerow(header)

    for row in filtered_rows:
      writer.writerow(row)

  print(f"Prodotti conservati in {csv_path}: {len(filtered_rows)}")


#------------------------------------------------------------------------------
# PARTE 4 - CONVERSIONE DA CSV A TSV
#------------------------------------------------------------------------------

def convert_csv_to_tsv(input_path, output_path):
  """
  Converte un file separato da virgole in un file separato da tabulazioni.

  Args:
    input_path (str): Percorso del file CSV originale.
    output_path (str): Percorso del file TSV da creare.
  """

  with open(input_path, "r", encoding="utf-8", newline="") as input_file:

    with open(output_path, "w", encoding="utf-8", newline="") as output_file:

      reader = csv.reader(input_file, delimiter=",")
      writer = csv.writer(output_file, delimiter="\t")

      # Copia anche l'intestazione, perché qui non viene saltata
      for row in reader:
        writer.writerow(row)

  print(f"File {output_path} creato.")


#------------------------------------------------------------------------------
# PARTE 5 - SCRITTURA E LETTURA DEL FILE JSON
#------------------------------------------------------------------------------

def create_and_read_inventory(json_path):
  """
  Salva un inventario in JSON, lo rilegge e stampa la prima taglia.

  Args:
    json_path (str): Percorso del file JSON da creare.
  """

  inventario = {
    "magliette": [
      {"taglia": "M", "colore": "rosso", "prezzo": 15.0},
      {"taglia": "L", "colore": "blu", "prezzo": 18.0}
    ],
    "felpe": [
      {"taglia": "S", "colore": "verde", "prezzo": 35.0}
    ],
    "pantaloni": [
      {"taglia": "XL", "colore": "nero", "prezzo": 40.0}
    ]
  }

  with open(json_path, "w", encoding="utf-8") as output_file:

    # Serializza il dizionario con indentazione e caratteri accentati leggibili
    json.dump(inventario, output_file, indent=2, ensure_ascii=False)

  with open(json_path, "r", encoding="utf-8") as input_file:
    loaded_inventory = json.load(input_file)

  # Convertendo un dizionario in lista si ottiene la lista delle sue chiavi
  categories = list(loaded_inventory)
  first_category = categories[0]

  # Accede alla categoria, al primo prodotto e infine al campo "taglia"
  first_size = loaded_inventory[first_category][0]["taglia"]

  print(f"Taglia del primo prodotto della prima categoria: {first_size}")


#==============================================================================
# PROGRAMMA PRINCIPALE
#==============================================================================

read_document("documento.txt")

write_and_append_text("output.txt")

filter_shirts("magliette.csv")

convert_csv_to_tsv("magliette.csv", "magliette.tsv")

create_and_read_inventory("inventario.json")
```
