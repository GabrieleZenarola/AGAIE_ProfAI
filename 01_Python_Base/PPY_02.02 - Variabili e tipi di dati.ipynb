## **Livello Principiante - 01**
Scrivi un programma Python che chieda all'utente di inserire il suo nome, la sua età e la sua altezza in metri (può essere un numero decimale). Il programma deve salvare questi dati in variabili separate e poi stampare un messaggio di saluto che includa tutte le informazioni inserite, formattate correttamente. Usa le funzioni input() e print(), assicurandoti di convertire i tipi di dato quando necessario. Infine, usa la funzione type() per stampare a schermo il tipo di dato di ciascuna variabile assegnata dall'input.

```python
name = input("Inserisci il tuo nome: ") #Richiede il nome con un form
age = int(input("Inserisci la tua età (numero intero): ")) #Richiede l'età con un form e la converte a intero
height = float(input("Inserisci la tua altezza in metri: ")) #Richiede l'altezza con un form e la converte a float

print()
# Uso di f-string per includere variabili nel messaggio, formato dell'altezza con 2 decimali
print(f"Ciao {name},\nhai {age} e sei alto {height:.2f}") #Stampa il messaggio, formattando l'altezza a 2 decimali
# Stampo i tipi delle variabili con la funzione type(), per capire il tipo di dato assegnato
print()
print(f"Tipo di 'name': {type(name)}")
print(f"Tipo di 'age': {type(age)}")
print(f"Tipo di 'height': {type(height)}")
```

```python
name = "Gabriele" # @param #Richiede il nome con un form
age = 50 # @param {type:"integer"} #Richiede l'età con un form e la converte a intero
height = 1.789 # @param {type: "number"} #Richiede l'altezza con un form e la converte a float

# Uso di f-string per includere variabili nel messaggio, formato dell'altezza con 2 decimali
print(f"Ciao {name},\nhai {age} anni e sei alto {height:.2f} m") #Stampa il messaggio, formattando l'altezza a 2 decimali
# Stampo i tipi delle variabili con la funzione type(), per capire il tipo di dato assegnato
print()
print(f"Tipo di 'name': {type(name)}")
print(f"Tipo di 'age': {type(age)}")
print(f"Tipo di 'height': {type(height)}")
```

## **Livello Intermedio - 01**
In questa esercitazione scriverai un programma che:

1) Chiede all'utente di inserire il proprio nome, la propria età (in anni) e l'altezza in metri (ad esempio 1.75).

2) Salva questi dati in variabili con nomi significativi.

3) Verifica e mostra il tipo di ciascuna variabile subito dopo l'assegnazione (usa print e type()).

4) Converte esplicitamente l'età in intero (int) e l'altezza in float utilizzando il casting, così da poter effettuare operazioni aritmetiche.

5) Calcola l'anno di nascita approssimativo sottraendo l'età dall'anno corrente (usa un valore fisso, ad esempio 2026).

6) Stampa un messaggio formattato che mostra il nome, l'anno di nascita e l'altezza formattata con due cifre decimali di precisione.

7) Gestisce i casi in cui l'input per età o altezza non sia un numero valido generando un errore (ad esempio se si prova a convertire una stringa non numerica in int o float, il programma deve fermarsi naturalmente senza righe di codice di gestione eccezioni che non sono ancora state studiate).

L'obiettivo è praticare l'acquisizione dei dati tramite input(), la loro memorizzazione in variabili, la verifica dei tipi con type(), il casting esplicito, l'uso di operazioni numeriche semplici e la stampa formattata usando f-string.

Usa i metodi di formattazione conosciuti per stampare i float con due decimali e mantieni nomi di variabili chiari e coerenti con le convenzioni Python (snake_case).

```python
#Richiesta dei dati di input: nome, età, altezza
user_name = input("Inserisci il tuo nome: ")
user_age = input("Inserisci la tua età: ")
user_height = input("Inserisci la tua altezza in metri: ")

#Stampa i tipi di dati prima del casting
print()
print("Stampa della variabili di input prima del casting")
print(f"Tipo di 'user_name': {type(user_name)}")
print(f"Tipo di 'user_age': {type(user_age)}")
print(f"Tipo di 'user_height': {type(user_height)}")

# Casting dell'input ('età' è un intero, 'altezza' è un decimale con 2 cifre significative) + gestione degli errori con ciclo 'if'

error = 0 #Variabile identificativa di un errore, se 0, allora è tutto OK, altrimenti c'è un errore di conversione

#Controllo che l'età sia un numero intero
if user_age.isdigit():
    user_age_int = int(user_age) #casting di età a intero
else:
    print()
    print("L'età deve essere un numero intero.")
    error = 1 #attivazione delle variabile 'errore'

#Controllo che l'altezza si un numero positivo
if user_height.replace('.', '', 1).replace('-', '', 1).isdigit():
    user_height_float = float(user_height) #casting dell'altezza in mumero decimale
    #Controllo che l'altezza sia un numero decimale positivo
    if user_height_float <= 0:
        print("L'altezza deve essere un numero positivo.")
        error = 1 #attivazione della variabile 'errore
else:
    print()
    print("L'altezza deve essere un numero decimale.")
    error = 1 #attivazione della variabile 'errore'

#Stampa dei messaggi finali, considerando il caso senza errori e con errori
if error == 0:
    #Calcolo dell'anno di nascita
    user_birth_year = 2024 - user_age_int
    #Stampa i tipi di dati dopo il casting
    print()
    print("Stampa della variabili dopo il casting")
    print(f"Tipo di 'user_name': {type(user_name)}")
    print(f"Tipo di 'user_age_int': {type(user_age_int)}")
    print(f"Tipo di 'user_height_float': {type(user_height_float)}")

    #Stampa il messaggio finale, se non ci sono errori
    print()
    print(f"{user_name}, nato approssimativamente nel {user_birth_year} è alto {user_height_float:.2f} metri")
else:
    #Stampa messaggio in caso di errori
    print()
    print("Si è verificato un errore durante l'inserimento dei dati.")

```

## **Livello Esperto - 01**
Realizza un programma Python che gestisca dati di diversi tipi di variabili e ne effettui conversioni, analisi di tipo e misurazione in memoria, rispettando rigorosamente la tipizzazione forte di Python.

Il programma dovrà:

1. Ricevere in input dalla tastiera tre valori distinti (ognuno su una riga): un numero intero, un numero decimale (float) e una stringa che rappresenta un numero intero.

2. Assegnare ciascun valore a una variabile distinta usando nomi significativi e seguendo la convenzione snake_case.

3. Effettuare il casting esplicito:
   - Convertire la stringa numerica in un intero.
   - Sommare il numero intero ottenuto dalla conversione con il primo valore intero.
   - Convertire infine la somma in float e sommarla con il numero decimale di partenza.

4. Stampare i risultati intermedi e finali, indicando chiaramente il tipo di ciascun valore usando la funzione type(), es. "Valore: 10, Tipo: <class 'int'>".

5. Calcolare e stampare la dimensione in byte (memoria occupata) di ogni variabile, usando sys.getsizeof (ricorda di importare sys).

6. Gestire correttamente eventuali errori di conversione con try-except, segnalando all’utente un messaggio chiaro (ad esempio se la stringa non rappresenta un numero intero).

7. Infine, acquisire un input libero (stringa) e formattare un output che mostri tutti i valori elaborati (intero, float, stringa, somma) in un’unica riga usando f-string, con la precisione del float limitata a due cifre decimali.

Nota bene: devi garantire che il programma non produca eccezioni non gestite e che rispetti la tipizzazione forte di Python senza concatenare direttamente tipi non compatibili senza conversione.

Questo esercizio ti permette di consolidare la gestione di variabili, tipi di dato dinamici, conversioni esplicite, gestione degli errori, e una formattazione avanzata degli output, tutti argomenti già affrontati nei moduli precedenti.


###**Esempi di Input/Output:**
input:
42
3.14159
17
hello <br>
output:
Valore 1: 42, Tipo: <class 'int'>, Dimensione in byte: 28
Valore 2: 3.14159, Tipo: <class 'float'>, Dimensione in byte: 24
Valore 3: '17', Tipo: <class 'str'>, Dimensione in byte: 50
Valore 3 convertito a int: 17, Tipo: <class 'int'>, Dimensione in byte: 28
Somma interi: 59, Tipo: <class 'int'>, Dimensione in byte: 28
Somma finale float: 62.14159, Tipo: <class 'float'>, Dimensione in byte: 24
Stringa libera: hello
Output formattato: Intero=59, Float=62.14, Stringa='hello'

input:
10
2.5
abc
pippo <br>
output:
Valore 1: 10, Tipo: <class 'int'>, Dimensione in byte: 28
Valore 2: 2.5, Tipo: <class 'float'>, Dimensione in byte: 24
Valore 3: 'abc', Tipo: <class 'str'>, Dimensione in byte: 52
Errore: la stringa 'abc' non è un numero intero valido.
Stringa libera: pippo
Output formattato: Intero=10, Float=2.50, Stringa='pippo'

```python
from sys import getsizeof #Importo la funzione getsizeof()

value_1_int = int(input("Inserisci il primo valore intero: ")) #Richiesta del primo valore intero
value_2_float = float(input("Inserisci il secondo valore decimale: ")) #Richiesta del secondo valore decimale
value_3_string = input("Inserisci il terzo valore intero: ") #Richiesta del terzo valore intero
string_free = input("Inserisci una stringa: ") #Richiesta di una stringa

#Realizzo il casting con gestione dell'errore
try:
    value_3_int = int(value_3_string) #Casting del terzo valore in intero
except ValueError:
    print(f"Errore: la stringa '{value_3_string}' non è un numero intero valido.")
    exit()  #Termina il programma in caso di errore per evitare eccezioni non gestite

#Realizzo la somma degli interi
sum_value_1_3 = value_1_int + value_3_int
# Conversione della somma a float e somma con valore float originale
sum_value_1_3_2_float = float(sum_value_1_3) + value_2_float

#Stampo i messaggi finali
print()
print(f"Valore 1: {value_1_int}, Tipo: {type(value_1_int)}, Dimensione in byte: {getsizeof(value_1_int)}")
print(f"Valore 2: {value_2_float:.2f}, Tipo: {type(value_2_float)}, Dimensione in byte: {getsizeof(value_2_float)}")
print(f"Valore 3: {value_3_string}, Tipo: {type(value_3_string)}, Dimensione in byte: {getsizeof(value_3_string)}")
print(f"Valore 3 convertito a int: {value_3_int}, Tipo: {type(value_3_int)}, Dimensione in byte: {getsizeof(value_3_int)}")
print(f"Somma interi: {sum_value_1_3}, Tipo: {type(sum_value_1_3)}, Dimensione in byte: {getsizeof(sum_value_1_3)}")
print(f"Somma finale float: {sum_value_1_3_2_float:.2f}, Tipo: {type(sum_value_1_3_2_float)}, Dimensione in byte: {getsizeof(sum_value_1_3_2_float)}")
print(f"Stringa libera: {string_free}")
print()
print(f"Output formattato: Intero = {sum_value_1_3}, Float = {sum_value_1_3_2_float:.2f}, Stinga = {string_free}")
```
