#**01-Principiante**
In questa esercitazione lavorerai con i cicli `while` e `for` per completare semplici compiti di programmazione. Scrivi programmi che utilizzano i cicli per ripetere operazioni come contare numeri, sommare valori e iterare su liste. Ricorda di inizializzare correttamente le variabili contatore e di incrementarle appropriatamente per evitare cicli infiniti. Usa print() per mostrare i risultati sullo schermo e input() per ricevere eventuali dati dall'utente.

1) Scrivi un programma che conta da 1 a un numero scelto dall'utente, stampando ogni numero su una riga.

2) Scrivi un programma che prende una lista di numeri (ad esempio [3, 5, 7, 9]) e calcola la somma di tutti gli elementi usando un ciclo for.

3) Scrivi un programma che chiede all'utente di inserire parole fino a quando digita la parola "fine". Usa un ciclo while per leggere le parole e termina il ciclo quando compare "fine", stampando quante parole sono state inserite (escludendo "fine").

##**Esempio di Input/Output**
input:
5 <br>
output:
1
2
3
4
5

input:
[3, 5, 7, 9] <br>
output:
24

input:
casa
cane
auto
fine <br>
output:
Hai inserito 3 parole.

```python
# STEP 1: Scrivo un programma che conta da 1 a un numero scelto dall'utente, stampando ogni numero su una riga.

# 1. Richiesta dell'input, con controllo dell'errore
while True:
  try:
    input_int = int(input("Inserisci un numero intero positivo: ")) # Richiesta input
    if input_int <= 0: # Controllo che il numero sia positivo
      print("Il numero deve essere positivo.")
    else:
      break # Uscita dal ciclo per continuare con lo step successivo
  except ValueError: # Messaggio d'errore se l'input non è un numero
    print("Input non valido. Inserisci un numero intero positivo.")

# 2. Ciclo 'for' per il conteggio
print("\nStampa del conteggio dei numeri con ciclo 'for'") # Messaggio di inizio
for i in range(1, input_int + 1):
  print(i) # Stampa del conteggio

print()

# 3. Ciclo 'while' per il conteggio
print("\nStampa del conteggio dei numeri con ciclo 'while'") # Messaggio di inizio
i = 0 # Inizializzazione del contatore
while i < input_int:
  print(i+1)
  i+=1

#=============================================================================

# STEP 2: Scrivo un programma che prende una lista di numeri (ad esempio [3, 5, 7, 9])
# e calcola la somma di tutti gli elementi usando un ciclo for.

input_list = [3, 5, 7, 9] # Lista in input

# Ciclo 'for' per sommare i numeri della lista
sum_number = 0 # Inizializzazione della somma
for num in input_list: # Il ciclo scorre tutte le posizioni della lista
  sum_number = sum_number + num

print(f"\nLa somma dei numeri della lista è: {sum_number}\n") # Stampa della somma

#=============================================================================

# STEP 3: Scrivo un programma che chiede all'utente di inserire parole fino a quando
# digita la parola "fine". Uso un ciclo while per leggere le parole e termina il ciclo
# quando compare "fine", stampando quante parole sono state inserite (escludendo "fine").

word_count = 0 # Inizializzazione del conteggio delle parole

# Ciclo 'while' per inserire le parole
while True:
  word = input("Inserisci la parola desiderata. 'Fine' per interrompere il ciclo: ")
  if word.lower() == "fine": # '.lower' trasforma l'input in minuscolo per confrontarlo con 'fine'
    break # Se l'input = fine, il ciclo temrina
  else:
    word_count += 1 # Incremento del contatore

print(f"Hai inserito {word_count} parole.") # Stampa del conteggio
```

#**02-Facile**
In questa esercitazione lavorerai con i cicli (loop) in Python utilizzando liste, variabili contatore e condizioni. Dato un elenco di numeri interi, ti viene chiesto di scrivere un programma che:

1. Calcoli la somma di tutti i numeri presenti nella lista.
2. Determini quanti numeri nella lista sono pari.
3. Stampi ciascun numero della lista con il rispettivo indice (a partire da 1).

Per completare l'esercizio dovrai utilizzare un ciclo for o while per iterare sulla lista, un contatore per conteggiare i numeri pari e operatori aritmetici e condizionali per testare la parità dei numeri. Ricorda di incrementare opportunamente i contatori e di utilizzare la funzione len() per ottenere la lunghezza della lista.

##**Esempio di Input/Output**
input:
[4, 7, 12, 17, 8] <br>
output:
Somma: 48
Numeri pari: 3
1: 4
2: 7
3: 12
4: 17
5: 8

input:
[10, 3, 5] <br>
output:
Somma: 18
Numeri pari: 1
1: 10
2: 3
3: 5

```python
# Definizione della lista in input
input_list = [4, 7, 12, 17, 8]

# 1. Calcolo della somma di tutti i numeri nella lista
sum_num = 0 # Inizializzazione della somma
for num in input_list: # Ciclo for per scorrere tutti i numeri della lista
  sum_num += num # Somma della lista

print(f"Somma: {sum_num}\n") # Stampa della somma

# 2. Determino quanti numeri nella lista sono pari.
count_even = 0 # Inizializzazione del contatore
for num in input_list: # Ciclo for per scorrere tutti i numeri della lista
  if num % 2 == 0: # Se il numero è pari
    count_even += 1 # Incremento del contatore

print(f"Numeri pari: {count_even}\n") # Stampa del conteggio

# 3. Stampo ciascun numero della lista con il rispettivo indice (a partire da 1).
for i, num in enumerate(input_list): # 'enumerate' trasforma la lista in una tupla (posizione, valore)
  print(f"{i+1}: {num}") # Stampa dell'elenco
```

#**03-Intermedio**
Scrivi un programma Python che:

1. Riceva in input dall'utente una lista di numeri interi, separati da spazi.
2. Utilizzi un ciclo for per iterare su ogni numero nella lista e calcolarne la somma totale.
3. Utilizzi un ciclo while per contare quanti numeri nella lista sono pari.
4. Durante l'iterazione, se il numero è negativo, il ciclo for deve interrompersi immediatamente (usando break).
5. Infine, stampa la somma calcolata e il numero di valori pari trovati fino al punto di interruzione.

Nota: il programma deve gestire l'input come una stringa, convertirla in una lista di numeri interi e poi applicare i cicli come richiesto. Utilizza solo gli argomenti già affrontati (input/output, cicli for e while, condizioni, operatori logici, break).

Esempio:
Se l'input è: 3 6 7 4 -1 10
Il ciclo for si bloccherà al numero -1, quindi la somma e il conteggio dei pari saranno calcolati solo sui numeri precedenti (3, 6, 7, 4).

Stampa:
Somma: 20
Numeri pari: 2

##**Esempio di Input/Output**
input:
3 6 7 4 -1 10 <br>
output:
Somma: 20
Numeri pari: 2

input:
1 3 5 7 9 <br>
output:
Somma: 25
Numeri pari: 0

input:
2 4 6 8 10 <br>
output:
Somma: 30
Numeri pari: 5

```python
# STEP 1: Richiesta dell'input, conversione e controllo

input_list_int = [] # Inizializzazione della lista
while True:
  input_list_str = input("Inserisci una serie di interi divisi da spazi: ") # Richiesta dell'input
  try:
      input_list_int = [int(val.strip()) for val in input_list_str.split(" ")] # Casting in intero di ciascun valore separato da virgola e senza spazi
      break # Uscita dal ciclo 'while' alla fine del casting
  except ValueError:
    print("Input non valido. Inserisci una serie di interi divisi da spazi.") # Messaggio d'errore

#===============================================================================

# STEP 2: Analisi della lista rispettando le condizioni a seguire.
# 1. Utilizzo un ciclo 'for' per iterare su ogni numero nella lista e calcolarne la somma totale.
# 2. Se il numero è negativo, il ciclo 'for' deve interrompersi immediatamente (usando break).

valid_list = [] # Inizializzazione della sottolista dei numeri pari
sum_num = 0 # Inizializzazione della somma
for num in input_list_int: # Analisi della lista completa
  if num < 0: # Se il numero è negativo, interrompe il ciclo
    break
  else:
    valid_list.append(num) # Aggiunge il numero alla lista dei numeri pari
    sum_num += num # Somma dei numeri positivo

print(f"Somma: {sum_num}") # Stampa della somma

#===============================================================================

# STEP 3: # Utilizzo un ciclo 'while' per contare i numeri pari della sottolista
count_even = 0 # Inizializzazione del conteggio pari
i = 0 # Inizializzazione del contatore/posizione del ciclo 'while'
while i < len(valid_list):
  if valid_list[i] % 2 == 0: # Se il numero è pari
    count_even += 1 # Incremento del conteggio
  i += 1 # Incremento del contatore

print(f"Numeri pari: {count_even}") # Stampa del conteggio
```

#**04-Difficile**
Scrivi un programma in Python che riceva in input una lista di numeri interi (fino a 20 elementi) inseriti uno alla volta dall'utente. Il programma deve eseguire le seguenti operazioni utilizzando esclusivamente cicli (while e for) e strutture dati apprese:

1. Prima di tutto, continua a leggere i numeri finché l’utente non inserisce il valore di fine input: il numero zero (0), che non deve essere considerato nel calcolo.

2. Calcola e stampa la somma di tutti i numeri inseriti.

3. Calcola e stampa la media aritmetica dei numeri inseriti.

4. Conta e stampa quanti numeri nella lista sono pari.

5. Stampa la lista degli elementi in ordine inverso (senza usare funzioni built-in di inversione o metodi non ancora appresi, ma con slicing o cicli).

6. Infine, chiedi all’utente di inserire un valore intero e verifica con un ciclo for se questo valore è presente nella lista. Stampa un messaggio adeguato di conferma o negazione.

Il programma deve essere scritto in modo da evitare cicli infiniti e deve fare un uso corretto di variabili contatori e condizioni di terminazione. Usare l’istruzione break o continue può essere utile, ma non obbligatorio.

Nota bene: non usare ancora funzioni, moduli esterni, né metodi avanzati di liste che non siano state affrontate fino ad ora.

##**Esempio di Input/Output**
input:
5
10
2
7
0 <br>
input per ricerca:
10 <br>
output:
Somma: 24
Media: 6.0
Numeri pari: 2
Lista invertita: [7, 2, 10, 5]
Il numero 10 è presente nella lista.

input:
4
3
12
1
0 <br>
input per ricerca:
8 <br>
output:
Somma: 20
Media: 5.0
Numeri pari: 2
Lista invertita: [1, 12, 3, 4]
Il numero 8 non è presente nella lista.

```python
# STEP 1: Ricevo in input una lista di numeri interi (fino a 20 elementi) inseriti uno alla volta dall'utente.
# Il valore di fine input è il numero zero (0), che non deve essere considerato nel calcolo.

input_list_int = [] # Inizializzazione della lista
i = 0 # Inizializzazione del contatore
while i < 20: # Ciclo while per inserire i numeri
  try:
    input_number = int(input("Inserisci un numero intero: ")) # Richiesta dell'input
    if input_number == 0: # Se l'input è 0, interrompe il ciclo
      break
    else:
      input_list_int.append(input_number) # Aggiunge il numero alla lista
      i += 1 # Incremento del contatore
  except ValueError:
    print("Input non valido. Inserisci un numero intero.") # Messaggio d'errore


print (input_list_int)

# STEP 2: Calcolo e stampo la somma di tutti i numeri inseriti.
sum_num = 0 # Inizializzo la variabile somma
for num in input_list_int:
  sum_num += num

print(f"Somma: {sum_num}") # Stampa della somma

# STEP 3: Calcolo e stampo la media aritmetica dei numeri inseriti.
# Aggiungo controllo: evito divisione per zero se la lista è vuota
if len(input_list_int) == 0:
  print("La lista è vuota. La media non può essere calcolata.")
else:
  mean_num = sum_num / len(input_list_int)
  print(f"Media: {mean_num}") # Stampa della media

# STEP 4: Conto e stampo quanti numeri nella lista sono pari.
count_even = 0 # Inizializzazione del conteggio pari
for num in input_list_int:
  if num % 2 == 0:
    count_even += 1

print(f"Numeri pari: {count_even}") # Stampa del conteggio

# STEP 5: Stampo la lista degli elementi in ordine inverso
inv_input_list = input_list_int[::-1] # Slicing della lista in ordine inverso
print(f"Lista invertita: {inv_input_list}") # Stampa della lista

# STEP 6: Chiedo all’utente di inserire un valore intero e verifica con un ciclo for se questo valore è presente nella lista.
while True:
  try:
    verification_number = int(input("Inserisci un numero da cercare nella lista: ")) # Richiesta dell'input
    # Uso il ciclo 'for' come richiesto dall'esercizio
    found_number = False # Inizializzazione della variabile di controllo
    for num in input_list_int:
      if num == verification_number:
        found_number = True # Se il numero è trovato, aggiorna la variabile di controllo
        break # Uscita dal ciclo 'for'
    if found_number:
      print(f"Il numero {verification_number} è presente nella lista.") # Stampa del messaggio di conferma
      break # Uscita dal ciclo 'while'
    else:
      print(f"Il numero {verification_number} non è presente nella lista.") # Stampa del messaggio di negazione
      break # Uscita dal ciclo 'while'
  except ValueError:
    print("Input non valido. Inserisci un numero intero") # Messaggio d'errore
```

#**05-Esperto**
Scrivi un programma Python che gestisca elaborazioni complesse su liste di dati numerici o stringhe sfruttando i costrutti dei cicli while e for, utilizzando anche break e continue.

Il programma dovrà:

1. Ricevere in input una lista di numeri interi o una lista di stringhe (potrai definire la lista direttamente nel codice, oppure riceverla tramite input con il casting opportuno).

2. Utilizzare un ciclo for combinato con enumerate per iterare sulla lista e stampare ciascun elemento preceduto dal suo indice, ma con queste regole:
  - Se l'elemento è un numero pari, continua senza stamparlo (usa continue).
  - Se l'elemento è una stringa che contiene la sottostringa "skip", salta la stampa (usa continue).
  - Se l'elemento è il valore speciale 0 (nel caso numerico) o la stringa "stop", ferma l'iterazione (break).

3. Implementa anche una versione con ciclo while per scorrere la lista fino a incontrare il valore speciale (0 o "stop"). Durante la scansione con while:
  - Somma i valori numerici (solo quelli dispari).
  - Conta quante stringhe conformi (cioè senza "skip" e prima di "stop") sono state incontrate.

4. Al termine di entrambe le iterazioni, stampa:
  - Il numero totale di elementi processati.
  - La somma calcolata (per la versione con while sui numeri).
  - Il conteggio degli elementi idonei (per la versione con while sulle stringhe).

Usa le tecniche di slicing, accesso tramite indici e iterazioni viste in teoria e pratica per gestire la lista in modo efficiente.

Non usare funzioni o concetti non ancora studiati (es. non definire funzioni, non usare lambda, non usare tipi complessi non inseriti prima).

##**Esempio di Input/Output**
input:
[1, 2, 3, 0, 5, 6] <br>
output:
Indice 0: 1
Indice 2: 3
Processati 4 elementi
Somma dispari con while: 4

input:
["ciao", "skipme", "hello", "stop", "world"] <br>
output:
Indice 0: ciao
Indice 2: hello
Processati 4 elementi
Conteggio stringhe idonee con while: 2

input:
[10, 12, 14, 0] <br>
output:
Processati 4 elementi
Somma dispari con while: 0

input:
["skip", "skipthis", "stop"] <br>
output:
Processati 3 elementi
Conteggio stringhe idonee con while: 0

```python
# STEP 1: Ricevo in input una lista di numeri interi o una lista di stringhe.
input_int_list = [1, 2, 3, 0, 5, 6] # Definizione della lista di interi
input_str_list = ["ciao", "skipme", "hello", "stop", "world"] # Definizione della lista di stringhe

# STEP 2: Utilizzo un ciclo for combinato con enumerate per iterare sulla lista e stampare
# ciascun elemento preceduto dal suo indice, ma con queste regole:
# 1. Se l'elemento è un numero pari, continuo senza stamparlo (usa continue).
# 2. Se l'elemento è una stringa che contiene la sottostringa "skip", salto la stampa (usa continue).
# 3. Se l'elemento è il valore speciale 0 (nel caso numerico) o la stringa "stop", fermo l'iterazione (break).

# Ciclo 'for' per la gestione della lista di interi:
for i, num in enumerate(input_int_list): # 'enumerate' crea un lista di tupe (indice, valore), in questo caso (i, num)
  if num % 2 == 0: # Se il numero è pari, salta l'iterazione
    continue
  elif num == 0: # Se il numero è 0, interrompe l'iterazione
    break
  else:
    print(f"Indice {i}: {num}") # Stampa dell'indice e del valore

# Ciclo 'for' per la gestione della lista di stringhe
for i, string in enumerate(input_str_list): # 'enumerate' crea un lista di tupe (indice, valore), in questo caso (i, string)
  if "skip" in string: # Se la stringa contiene 'skip', salta l'iterazione
    continue
  elif string == "stop": # Se la stringa è 'stop', interrompe l'iterazione
    break
  else:
    print(f"Indice {i}: {string}") # Stampa dell'indice e della stringa

# STEP 3: Implemento anche una versione con ciclo while per scorrere la lista fino a
# incontrare il valore speciale (0 o "stop"). Durante la scansione con while:
# 1. Sommo i valori numerici (solo quelli dispari).
# 2. Conto quante stringhe conformi (cioè senza "skip" e prima di "stop") sono state incontrate.

# Ciclo 'while' per la lista di interi:
sum_num = 0 # Inizializzazione della somma
i = 0 # Inizializzazione del contatore
while i < len(input_int_list):
  if input_int_list[i] == 0: # Se il numero è 0, interrompe l'iterazione
    break
  elif input_int_list[i] % 2 != 0: # Se il numero è dispari, somma il valore alla somma
    sum_num += input_int_list[i]
  i += 1 # Incremento del contatore

print(f"Processati {i+1} elementi") # Stampa del conteggio
print(f"Somma dispari con while: {sum_num}") # Stampa della somma

# Ciclo 'while' per la lista di stringhe:
right_str_count = 0 # Inizializzo il contatore delle stringhe idonee
c = 0 # Inizializzazione del contatore
while c < len(input_str_list):
  if input_str_list[c] == "stop": # Se la stringa è 'stop', interrompe l'iterazione
    break
  elif "skip" in input_str_list[c]: # Se 'skip' è parte della stringa, salta l'iterazione
    c += 1
    continue
  else: # Se 'skip' non è parte della stringa, si incrementa il contatore delle stringhe idonee
    right_str_count += 1
  c += 1 # Incremento del contatore del ciclo 'while'

print(f"Processati {c+1} elementi") # Stampa del conteggio
print(f"Conteggio stringhe idonee con while: {right_str_count}")
```
