## **1. Area della Circonferenza**
Definisci una funzione che, prendendo in input il raggio, restituisce l'area della circonferenza.
<br><br>
**SUGGERIMENTI**<br>
Puoi calcolare l'area della circonferenza come $ A = \pi r^2 $ dove $ \pi $ è il pi greco e può essere approssimato a 3,14.

```python
# Importo il modulo math e definisco pi
import math
pi = math.pi

# Definizione della funzine per calcolare l'area della circonferenza, dato il raggio
def circle_area (radius):
  circle_area = round(pi * (radius**2), 2)

  return circle_area

# Richiesta del raggio in input + Gestione errore
while True:
  try:
    radius = round(float(input("Inserisci il raggio della circonferenza: ")), 2)
    break
  except ValueError:
    print("Non hai inserito un numero valido")

# Invocazione della funzione per calcolare l'area
area = circle_area(radius)

# Stampa del risultato
print(f"L'area della circonferenza di raggio {radius} è: {area}")
```

## **2. Anni bisestili**
Ogni 4 anni aggiungiamo un giorno al mese più breve dell'anno, cioè Febbraio, quest'anno è chiamato bisestile. I criteri utilizzati per definire se un anno è bisestile sono i seguenti:
* L'anno è divisibile per 4, allora è bisestile a meno che:
* * L'anno è divisibile per 100, allora non è bisestile a meno che:
* * * L'anno è divisile per 400.

Scrivi una funzione per verificare se un anno è bisestile.

```python
# Definisco la funzione per determinare se l'anno inserito è bisestile
def is_leap_year (year):
  is_leap_year = False # Variabile booleana per stabilire se l'anno è bisestile
  if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0 and year % 100 != 0):
    is_leap_year = True

  return is_leap_year

# Richiesta dell'anno come input (dev'essere un numero intero)
while True:
  try:
    year = int(input("Inserisci un anno: "))
    break
  except ValueError:
    print("Non hai inserito un numero valido")

# Invocazione della funzione per determinare se l'anno è bisestile
is_leap = is_leap_year(year)

# Stampo il risultato
print(f"L'anno {year} " + ("" if is_leap else "non ") + "è bisestile")

```

## **3. Anagrammi**
Definisci una funzione che, prendendo in input due parole, verifica se le due parole sono anagrammi.
<br><br>
**SUGGERIMENTO**
<br>
Due parole sono anagrammi se sono composte esattamente dalle stesse lettere (es. strani/nastri, carino/canori, senatori/estranio)

```python
# Definisco una funzione che riceve 2 parole come argomenti e le confronta
# per stabilire se sono anagrammi
def is_anagram (word1, word2):
  is_anagram = True # Variabile booleana per stabilire se le due parole sono anagrammi

  # Se le due parole hanno diversa lunghezza, allora non sono anagrammi
  # Questo controllo è utile per parole lunghe: evita cicli inutili
  if len(word1) != len (word2):
    return False # Restituisce il valore booleano ed esce dalla funzione

  # Se le due parole hanno uguale lunghezza, allora provo a rimuovere ciascuna
  # lettera della prima parola, dalla seconda
  for letter in word1:
    if letter in word2:
      # Se esiste una corrispondenza di lettere, la rimuove dalla seconda parola
      word2 = word2.replace(letter, "", 1)
    else:
      # Se una lettera non corrisponde, non è un anagramma
      return False # Restituisce il valore booleano ed esce dalla funzione

  return is_anagram # Restituisce il valore booleano alla fine del ciclo 'for'


# Richiesta delle parole in input
word1 = input("Inserisci la prima parola: ")
word2 = input("Inserisci la seconda parola: ")

# Invocazione della funzione
is_anagram_result = is_anagram(word1, word2)

# Stampo il risultato
print(f"Le parole {word1} e {word2} " + ("" if is_anagram_result else "non ") + "sono anagrammi")

```

## **4. Validazione di indirizzi email**
Realizziamo una funzione che, prendendo in input una lista di indirizzi email, ritorni una nuova lista degli indirizzi email validi ordinati alfabeticamente. Un'indirizzo email è valido se:
* Ha il formato nomeutente@dominio.estensione.
* Il nome utente contiene soltanto lettere, numeri, lineette (-) e trattini bassi (_).
* Il dominio contiene soltanto lettere e numeri.
* La lunghezza massima dell'estensione è 3 caratteri.

```python
# Definisco la funzione di validazione degli indirizzi e-mail che riceve un lista
# di e-amil come argomento
def email_validator(email_list):
  valid_emails = [] # Lista vuota per memorizzare gli e-mail validi
  temporary_list = [] # Lista temporanea per memorizzare gli e-mail temporaneamente
  invalid_emails = [] # Lista vuota per memorizzare gli e-mail non validi

  # Controlla che esista un solo carattere "@"", un solo "punto" per ogni
  # indirizzo e-mail della lista e che il punto sia dopo "@".
  # In caso contrario, popola la lista di e-mail
  # non valide e le elimina dalla lista di e-mail in input
  for email in email_list:
    if email.count("@") != 1 or email.count(".") != 1:
      invalid_emails.append(email) # Popola la lista di e-mail non valide
    elif email.index(".") < email.index("@"):
      invalid_emails.append(email) # Popola la lista di e-mail non valide
    else:
      temporary_list.append(email) # Popola la lista temporanea con gli e-mail validi

  # Divide le mail con un solo "@" ed un solo "punto" in "username", "domain",
  # "extension", secondo le regole seguenti:
  # 1. Lo "username" è la stringa che precedere il simbolo "@"
  # 2. Il "domain" è compreso tra "@" ed il "punto"
  # 3. La "extension" è la stringa successiva al "punto"

  correct_user_list = [] # Lista vuota per memorizzare gli username validi
  correct_domain_list = [] # Lista vuota per memorizzare i domini validi

  # 1. Controlla che il nome utente contienga soltanto lettere, numeri,
  # lineette (-) e trattini bassi (_)
  for email in temporary_list:
    username = email.split("@")[0] # Identifico il nome utente
    valid_user = bool(username) # Variabile booleana per stabilire se il nome utente è valido

    for char_user in username:
      if char_user.lower() not in "abcdefghijklmnopqrstuvwxyz0123456789-_":
        valid_user = False
        break # Interrompe il ciclo 'for' se uno dei caratteri non è valido

    if valid_user:
      correct_user_list.append(email) # Aggiunge il carattere alla lista di caratteri validi
    else:
      invalid_emails.append(email) # Aggiunge il carattere alla lista di caratteri non validi

  # 2. Controlla che il dominio contenga soltanto lettere e numeri
  for email in correct_user_list:
    domain = email.split("@")[1].split(".")[0] # Identifico il dominio
    valid_domain = bool(domain) # Variabile booleana per stabilire se il dominio è valido

    for char_dom in domain:
      if char_dom.lower() not in "abcdefghijklmnopqrstuvwxyz0123456789":
        valid_domain = False
        break # Interrompe il ciclo 'for' se uno dei caratteri non è valido

    if valid_domain:
      correct_domain_list.append(email) # Aggiunge il carattere alla lista di caratteri validi
    else:
      invalid_emails.append(email) # Aggiunge il carattere alla lista di caratteri non validi

  # 3. Controlla che l'estensione abbia lunghezza massima di 3 caratteri
  for email in correct_domain_list:
    extension = email.split("@")[1].split(".")[1] # Identifico l'estensione

    if 1 <= len(extension) <= 3:
      valid_emails.append(email) # Aggiunge l'e-mail alla lista di e-mail valide
    else:
      invalid_emails.append(email)

  valid_emails = sorted(valid_emails) # Ordina gli indirizzi validi in ordine alfabetico
  invalid_emails = sorted(invalid_emails) # Ordina gli indirizzi non validi in ordine alfabetico

  return valid_emails, invalid_emails # Restituisce le due liste di e-mail


# Richiesta di una lista di mail come input, finchè viene digitata la parola
# "stop" che non verrà memorizzata nella lista
email_list = [] # Lista vuota per memorizzare le mail inserite dall'utente
while True:
  email = input("Inserisci un indirizzo email (o 'stop' per terminare): ")
  if email.lower() == "stop":
    break
  email_list.append(email)

# Invoca la funzione
valid_emails, invalid_emails = email_validator(email_list)

# Stampa i risultati
print(f"Indirizzi email validi: {valid_emails}")
print(f"Indirizzi email non validi: {invalid_emails}")

```

```python
# Definizione della funzione per controllare che le mail siano state
# scritte correttamente
def email_validator(email_list):

  valid_emails = [] # Lista vuota per memorizzare gli e-mail validi
  invalid_emails = [] # Lista vuota per memorizzare gli e-mail non validi

  for email in email_list:

    # STEP 1: Controlla che l'indirizzo e-mail sia strutturato correttamente:
    # 1. Controlla che esista "@" e sia unico
    section_1 = email.split("@") # Divide la lista in 2 sezioni
    if len(section_1) != 2:
      invalid_emails.append(email) # Memorizza l'indirizzo e-mail non valido
      continue # Prosegue con la prossima e-mail della lista

    # 2. Controlla che esista il "." e sia dopo "@"
    section_2 = section_1[1].split(".") # Divide la seconda sezione in 2 sezioni
    if len(section_2) != 2:
      invalid_emails.append(email) # Memorizza l'indirizzo e-mail non valido
      continue # Prosegue con la prossima e-mail della lista

    # STEP 2: Controlla che il nome utente contenga solo caratteri alfanumerici
    # lineette (-) e trattini bassi (_).
    username = section_1[0] # Identifico il nome utente
    # Se eliminando "-" e "_" la stringa non contiente unicamente caratteri
    # alfanumerici, allora il nome è mal scritto
    if not username.replace("-", "").replace("_", "").isalnum():
      invalid_emails.append(email) # Memorizza l'indirizzo e-mail non valido
      continue # Prosegue con la prossima e-mail della lista

    # STEP 3: Controlla che il dominio sia composto unicamente da caratteri
    # alfanumerici
    domain = section_2[0] # Identifico il dominio
    if not domain.isalnum():
      invalid_emails.append(email) # Memorizza l'indirizzo e-mail non valido
      continue # Prosegue con la prossima e-mail della lista

    # STEP 4: Controlla che l'estensione abbia un lunghezza massima di 3
    # caratteri
    extension = section_2[1] # Identifico l'estensione
    if 1<= len(extension) <= 3:
      valid_emails.append(email) # Memorizza l'indirizzo e-mail valido
    else:
      invalid_emails.append(email) # Memorizza l'indirizzo e-mail non valido

  # Ordina le liste in ordine alfabetico
  valid_emails.sort()
  invalid_emails.sort()
  return valid_emails, invalid_emails # Ordina le liste in ordine alfabetico


# Invoca la funzione 'email_validator'
email_list = [
    "a@g.com",
    "abc_12@g.it",
    "@g.com",
    "a@.com",
    "a@g.",
    "a@g.comi",
    "a@g_.com",
    "a.b@g",
    "a@@g.com"
]

valid_emails, invalid_emails = email_validator(email_list)

# Stampa i risultati
print(f"Indirizzi email validi: {valid_emails}")
print(f"Indirizzi email non validi: {invalid_emails}")

```
