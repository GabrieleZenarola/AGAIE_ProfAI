## 1. La Circonferenza
Definisci una classe per rappresentare una circonferenza che, prendendo in input il raggio, definisca dei metodi per il calcolo di diametro, circonferenza e area. Documenta i metodi utilizzando le Docstring.

```python
# Definizione della classe 'Circle' dove, dato il raggio, calcola il diametro,
# la circonferenza e l'area.

# Per prima cosa si importa la funzione pi()
from math import pi

# Definizione della classe 'Circle', che avrà un unico attributo: il raggio 'r'
class Circle (): # Classe padre, quindi senza argomenti

  """
  Classe per rappresentare una circonferenza.
  """
  def __init__ (self, r): # Definizione del 'costruttore'
    """
    Costruttore della classe che riceve in input il raggio 'r'
    """
    self.r = r # Attributo della classe 'Circle'

  def diameter (self): # Definizione del metodo 'diameter'
    """
    Metodo che calcola il diametro della circonferenza
    """
    self.d = 2 * self.r # Calcolo del diametro
    return self.d # Restituzione del valore calcolato

  def perimeter (self): # Definizione del metodo 'perimeter'
    """
    Metodo che calcola la circonferenza
    """
    self.c = 2 * pi * self.r # Calcolo della circonferenza
    return self.c # Restituzione del valore calcolato

  def area (self): # Definizione del metodo 'area'
    """
    Metodo che calcola l'area della circonaferenza
    """
    self.a = pi * self.r ** 2 # Calcolo dell'area
    return self.a # Restituzione del valore calcolato

  def __repr__ (self): # Definizione del metodo '__repr__'
    """
    Metodo che restituisce una stringa con il diametro, la circonferenza e l'area della circonferenza
    """
    return (
        f"Il diametro della circonferenza è {self.diameter()}, "
        f"la circonferenza è {self.perimeter():.2f} "
        f"e l'area è {self.area():.2f}"
    )

# Richiamo il metodo 'help' per verificare l'uso delle 'docstrings'
help(Circle)

# Istanziazione della classe 'circle'
circle = Circle(3)
print(circle)
```

## 2. Vettori
Crea una classe che rappresenti un vettore numerico. La classe accetta in input una lista di numeri che saranno i valori del vettore. La classe deve avere queste funzionalità:
1. Stampando un'oggetto Vector il risultato sarà la lista dei suoi valori.
2. Sommando due oggetti Vector il risultato sarà un nuovo oggetto Vector composto dalla somma dei valori dei due vettori alle posizioni corrispondenti (es. Vector([1,2,3]) + Vector([2,4,6]) = Vector([3,6,9]).
3. Sottraendo un'oggetto Vector ad un'altro oggetto Vector il risultato sarà un nuovo oggetto Vector composto dalla differenza dei valori dei due vettori alle posizioni corrispondenti Vector([2,4,6]) + Vector([1,2,3]) = Vector([1,2,3]).
4. Moltiplicando due oggetti Vector il risultato dovrà essere il prodotto scalare dei due vettori, il prodotto scalare è definito come la somma del prodotto dei singoli elementi corrispondenti dei due vettori (es. Vector([1,2,3]) + Vector([2,4,6]) = 28.

Per i punti 2,3 e 4 i vettori devono avere uguale dimensione, in caso di dimensioni differenti stampa "I vettori hanno dimensione differente" e ritorna None.

Inoltre la classe deve avere i seguenti metodi:
1. **.sum()**: ritorna la somma di tutti gli elementi del vettore.
2. **.norm()**: ritorna la norma del vettore, cioè la radice quadrata del prodotto scalare tra il vettore e se stesso (es. Vector([10, 20, 30, 40]).norm() = 54.77)


Inoltre la classe deve supportare l'indexing <br><br>

```python
#==============================================================================
# DEFINIZIONE DELLE FUNZIONI E DELLE CLASSI
#==============================================================================

# Funzione per il casting a float della lista numerica in input
def check_list(number_list):
  number_list = number_list.split() # Split della stringa in input
  float_list = [] # Inizializzazione della lista numerica

  for val in number_list: # Scorre la lista in input
    try:
      val = float(val) # Casting dell'input in decimale
      float_list.append(val) # Aggiunge il valore alla lista numerica
    except ValueError: # Gestione dell'errore del casting a decimale
      print(f"Il valore '{val}' inserito non è un numero")
      return False # Ritorna False
  return float_list # Se il casting a decimale è corretto, ritorna la lista numerica


# Classe padre 'Vector'
class Vector(): # Classe padre, per tanto senza argomenti
  """
  Classe per rappresentare un vettore numerico.
  """

  def __init__(self, vector): # Definizione del 'costruttore'
    """
    Costruttore della classe che riceve in input una lista di numeri
    """
    self.vector = vector # Attributo della classe 'Vector'
    self.dim = len(vector) # Dimensione del vettore

  def __getitem__(self, i): # Definizione del metodo per prelevare un elemento del vettore
    """
    Metodo per prelevare un elemento specifico del vettore
    """
    return self.vector[i] # Restituzione dell'elemento


  def is_len(self, vector): # Definizione del metodo per stabilire se i vettori hanno la stessa lunghezza
    """
    Metodo per verificare se i vettori hanno la stessa lunghezza
    """
    if len(self.vector) != len(vector.vector): # Check lunghezze
      print("I vettori hanno dimensione differente")
      return False # Ritorna False
    return True # Ritorna True


  def __add__(self,vector): # Definizione del metodo somma vettoriale
    """
    Metodo per sommare due vettori
    """
    if self.is_len(vector): # Check lunghezze
      sum_list = []
      for i in range(len(self.vector)):
        sum_list.append(self.vector[i] + vector.vector[i])
      return Vector(sum_list)

  def __sub__(self,vector): # Definizione metodo sottrazione vettoriale

    """
    Metodo per sottrarre due vettori
    """
    if self.is_len(vector): # Check lunghezze
      dif_list = []
      for i in range(len(self.vector)):
        dif_list.append(self.vector[i] - vector.vector[i])
      return Vector(dif_list)

    return None

  def __mul__(self,vector): # Definizione metodo prodotto scalare
    """
    Metodo per moltiplicare due vettori
    """
    if self.is_len(vector): # Check lunghezze
      val = 0
      for i in range(len(self.vector)):
        val += self.vector[i] * vector.vector[i]
      return val

    return None

  def __repr__(self): # Definizione del metodo '__repr__'
    """
    Metodo che restituisce una stringa con i valori del vettore
    """
    return str(self.vector)

  def sum(self): # Definizione del metodo 'sum'
    """
    Metodo che calcola la somma di tutti gli elementi del vettore
    """
    sum = 0
    for val in self.vector:
      sum += val
    return sum

  def norm(self): # Definizione del metodo 'norm'
    """
    Metodo che calcola la norma del vettore
    """
    norm = 0
    for val in self.vector:
      norm += val ** 2
    return f"{float(norm**0.5):.2f}"



#==============================================================================
# CODICE PRINCIPALE
#==============================================================================

# Richiesta dei vettori di input
vector_input_1 = input("Inserisci la prima lista di numeri separati da uno spazio: ")
vector_input_2 = input("Inserisci la seconda lista di numeri separati da uno spazio: ")

# Assegnazione della lista di numeri a ciascun vettore
vector1 = check_list(vector_input_1) # Casting della prima lista in input
vector2 = check_list(vector_input_2) # Casting della seconda lista in input

if vector1 and vector2: # Se entrambi i vettori sono validi, si prosegue
  # Istanziamento della classe 'Vector' con i due vettori
  v1 = Vector(vector1)
  v2 = Vector(vector2)

  #Stampa dei risultati
  print(v1 + v2)
  print(v1 - v2)
  print(v1 * v2)
  print(v1.sum())
  print(v1.norm())
  print(v2.sum())
  print(v2.norm())
```
