# **1. Vettori Reloaded**
Durante la nostra esercitazione con le classi abbiamo realizzato una classe per rappresentare vettori, in più metodi abbiamo verificato che la lunghezza di due vettori fosse uguale. Rimpiazziamo l'if utilizzato in questi casi con un assert.

```python
class Vector():

    def __init__(self, values):
        self.values = values


    def __getitem__(self, i):
      return self.values[i]


    #Aggiungo il metodo __len__ che calcola la lunghezza del vettore
    """
    Metodo che calcola la lunghezza del vettore
    return:
      Lunghezza del vettore
    """
    def __len__(self):
      return len(self.values)

    def __repr__(self) :
       return str(self.values)


    def __add__(self, vector):

      # Sistema originale per il controllo delle lunghezze con ciclo 'if':
      if(len(self.values)!=len(vector.values)):
        print("I vettori hanno dimensioni diverse")
        return None

      # Sistema di controllo delle lunghezze attraverso il metodo 'assert'
      # La forma generale è: assert condizione, "messaggio"
      # Se è falsa, Python solleva un’eccezione AssertionError
      assert len(self)==len(vector), "I vettori hanno dimensioni diverse"
      list_sum = []

      for val1, val2 in zip(self.values, vector.values):
        list_sum.append(val1+val2)

      return Vector(list_sum)


    def __sub__(self, vector):

      # Sistema di controllo delle lunghezze attraverso il metodo 'assert'
      # La forma generale è: assert condizione, "messaggio"
      # Se è falsa, Python solleva un’eccezione AssertionError
      assert len(self)==len(vector), "I vettori hanno dimensioni diverse"

      list_sum = []

      for val1, val2 in zip(self.values, vector.values):
        list_sum.append(val1-val2)

      return Vector(list_sum)


    def __eq__(self, vector):

      # Sistema di controllo delle lunghezze attraverso il metodo 'assert'
      # La forma generale è: assert condizione, "messaggio"
      # Se è falsa, Python solleva un’eccezione AssertionError
      assert len(self)==len(vector), "I vettori hanno dimensioni diverse"

      for val1, val2 in zip(self.values, vector.values):
        if(val1 != val2):
          return False
      return True


    def __mul__(self, vector):
      return self.dot(vector)


    def sum(self):

      v_sum = 0

      for v in self.values:
        v_sum+=v

      return v_sum


    def dot(self, vector):

      # Sistema di controllo delle lunghezze attraverso il metodo 'assert'
      # La forma generale è: assert condizione, "messaggio"
      # Se è falsa, Python solleva un’eccezione AssertionError
      assert len(self)==len(vector), "I vettori hanno dimensioni diverse"

      prod = 0

      for val1, val2 in zip(self.values, vector.values):
        prod+=val1*val2

      return prod


    def norm(self):
      return self.dot(self)**(1/2)



v1 = Vector([1, 2, 3])
v2 = Vector([2, 4, 6, 7])

try:
  v1+v2
except Exception as e:
  print(e)
```

# **2. Calcolatrice**
Definisci una funzione per eseguire dei semplici calcoli, la funzione prende in ingresso due valori numerici ed una stringa contenente un'operazione aritmetica (+, -, *, /) ed esegue tale operazione tra i due numeri. Acquisisci in input i parametri della funzione
1. utilizza il try/except per gestire l'eccezione che si causa nel caso in cui viene inserito un valore non numerico.
2. utilizza l'assert per assicurarti che l'operatore aritmetico sia valido, cioè sia un carattere tra +,-,*,/.
3. utilizza il try/except per gestire il caso di divisioni per 0, in tal caso stampa semplicemente un messaggio 'Non puoi dividere per 0'.

Nel punto 1, il programma non deve interrompersi in casi di eccezione, ma deve continuare a chiedere i valori fino a quando questi non saranno corretti.

```python
#===============================================================================
# DEFINIZIONE DELLE FUNZIONI
#===============================================================================

# Funzione per la richiesta di input
def get_input():
  val = input("Inserisci un valore numerico: ")
  return val

# Funzione per il casting dell'input a float, con gestione dell'errore
def val_float(val):
  while True:
    try:
      return float(val)
    except ValueError:
      print("Il valore inserito non è un numero")
      val = input("Inserisci un valore numerico: ")

# Funzione per l'acquisizione dell'operatore matematico
def get_operation():
  operation = input("Inserisci un operatore (+,-,*,/): ")
  return operation

# Funzine per il controllo dell'operatore matematico
def math_operation(operation):
  while True:
    try:
      assert operation in ["+","-","*","/"], "L'operatore inserito non è valido"
      return operation
    except AssertionError as e:
      print(e)
      operation = get_operation()

# Funzione que esegue il calcolo matematico
def calc(val1, val2, operation):
  if operation == "+":
    return val1+val2
  elif operation == "-":
    return val1-val2
  elif operation == "*":
    return val1*val2
  elif operation == "/":
    try:
      return val1/val2
    except ZeroDivisionError:
      print("Non puoi dividere per 0")
      return None


#===============================================================================
# CODICE PRINCIPALE
#===============================================================================

# Richiesta dei valori di input
val1 = val_float(get_input())
val2 = val_float(get_input())

# Richiesta dell'operatore matematico
operation = math_operation(get_operation())

# Calcolo e stampa del risultato
result = calc(val1, val2, operation)
if result != None:
  print(f"Il risultato è: {result}")
```
