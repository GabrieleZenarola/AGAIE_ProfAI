4. Crea un set contenente solo le iniziali uniche dei nomi degli studenti.

- La lista senza duplicati ordinata.

Usa l'input() per leggere i dati come una singola riga di nomi separati da spazi, e processali opportunamente.

Attenzione: non modificare la lista originale direttamente durante la rimozione dei duplicati, ma costruisci una nuova lista pulita.

###**Esempio di Input/Output**

input:
Alice Bob Carla Alice Daniel Bob Eva <br>
output:
Lista senza duplicati ordinata: ['Alice', 'Bob', 'Carla', 'Daniel', 'Eva']
Tupla con nomi > 5 caratteri: ('Carla', 'Daniel')

input:
Luca Marco Luca Marco Francesca Elisa <br>
output:
Lista senza duplicati ordinata: ['Elisa', 'Francesca', 'Luca', 'Marco']
Tupla con nomi > 5 caratteri: ('Francesca',)
Set delle iniziali uniche: {'E', 'F', 'L', 'M'}
Dizionario con conteggio occorrenze: {'Luca': 2, 'Marco': 2, 'Francesca': 1, 'Elisa': 1}

```python
name_list=[] # Creo una lista vuota da popolare con i nomi

# Ciclo per popolare la lista con una serie di nomi, fino al comando "END"
print("Inserisci una serie di nomi. Termina l'inserimento scrivendo il comando 'END'")
while True:
    name = input("Inserisci un nome: ")
    if name == "END": # Intercettiamo la condizione di uscita prima di alterare i dati
        break
    name_list.append(name) # Popolo la lista
print(f"Lista di input generata: {name_list}") # Stampo la lista iniziale con i nomi

# Creo una lista pulita senza duplicati e ordinata alfabeticamente
name_set = set(name_list) # Casting della lista iniziale in Set per eliminare i duplicati
name_list_unique = list(name_set) # Casting del Set in Lista senza duplicati
name_list_unique.sort() # Ordinamento alfabetico della Lista senza duplicati
print(f"Lista senza duplicati ordinata: {name_list_unique}") # Stampa della Lista ordinata e senza duplicati

# Creo la Tupla con nomi più lunghi di 5 caratteri a partire da una lista che verrà popolata dai nomi lunghi
name_list_long =[] # Creo una lista vuota che verrà popolata con nomi con più di 5 caratteri
# Ciclo che scorre la lista di nomi unici, cercando quelli con più di 5 caratteri
for i in name_list_unique:
  if len(i) > 5:
    name_list_long.append(i) # Popolo la lista
name_tuple_long = tuple(name_list_long) # Casting della lista in Tupla
print(f"Tupla con nomi > 5 caratteri: {name_tuple_long}") # Stampo la tupla finale con nomi con più di 5 caratteri

# Creo il Set con le sole iniziali dei nomi unici a partire da un lista
name_first_letter_list = [] # Creo una lista vuoto
for i in name_list_unique:
  name_first_letter_list.append(i[0]) # Isolo la prima lettere di ogni nome
name_first_letter_set = set(name_first_letter_list) # Casting della lista in Set
print(f"Set delle iniziali uniche: {name_first_letter_set}") # Stampo il Set con le iniziali dei nomi unici

# Creo il Dizionario con la Chiave = "Nomi" in input e "Valore" = volte che si ripete il nome nell'input iniziale
name_dict_occurrences = {} # Creo un dizionario vuoto da popolare
for name in name_list: # Il ciclo itera direttamente sugli elementi della lista
    name_dict_occurrences[name] = name_list.count(name)
print(f"Dizionario con conteggio occorrenze: {name_dict_occurrences}") # Stampo il Dizionario con conteggio occorrenze
```

   Esempio: [
     ("Luca", "Rossi", [("Matematica", 28), ("Fisica", 30)]),
     ("Anna", "Bianchi", [("Matematica", 25), ("Informatica", 27)]),
     ("Marco", "Verdi", [("Fisica", 22), ("Informatica", 29)])
   ]

2. Creare un dizionario che abbia come chiave il nome del corso e come valore una lista di voti ottenuti in quel corso da tutti gli studenti.

3. Per ogni corso, calcolare la media aritmetica dei voti (arrotondata a due decimali).

4. Determinare e stampare il corso con la media voto più alta.

5. Creare un nuovo set contenente i nomi unici degli studenti che hanno preso almeno un voto superiore o uguale a 28.

6. Infine, stampare il numero totale di studenti iscritti (input originale) e il numero di studenti con voti >= 28.

Devi utilizzare esclusivamente le tecniche di indicizzazione, slicing, operazioni su liste, tuple, set e dizionari che hai già studiato. Ricorda che le tuple sono immutabili mentre le liste lo sono; le liste annidate possono essere iterate con doppio indexing. Non è richiesta la gestione di condizioni o try-except perché non sono stati ancora studiati, quindi assicurati che l’input sia sempre in formato corretto.

Realizza il programma in modo modulare (anche se non hai ancora definito funzioni, puoi comporre il codice per step chiari), con commenti per ogni passo.

###**Esempio di Input/Output**

input:
[("Luca", "Rossi", [("Matematica", 28), ("Fisica", 30)]), ("Anna", "Bianchi", [("Matematica", 25), ("Informatica", 27)]), ("Marco", "Verdi", [("Fisica", 22), ("Informatica", 29)])] <br>
output:
Corso con la media più alta: Fisica (media: 26.0)
Studenti totali: 3
Studenti con almeno un voto >= 28: 2

input:
[("Giulia", "Neri", [("Storia", 30), ("Italiano", 29)]), ("Paolo", "Gallo", [("Storia", 25), ("Italiano", 26)]), ("Elena", "Russo", [("Storia", 23)])] <br>
output:
Corso con la media più alta: Italiano (media: 27.5)
Studenti totali: 3
Studenti con almeno un voto >= 28: 1

input:
[("Mario", "Lombardi", [("Biologia", 24), ("Chimica", 27)]), ("Sara", "Marini", [("Biologia", 26)])] <br>
output:
Corso con la media più alta: Chimica (media: 27.0)
Studenti totali: 2
Studenti con almeno un voto >= 28: 0

```python
# STEP 1: Ricevo in input una lista annidata contenente dati di studenti iscritti a più corsi universitari
grades_list = [
    ("Luca", "Rossi", [("Matematica", 28), ("Fisica", 30)]),
    ("Anna", "Bianchi", [("Matematica", 25), ("Informatica", 27)]),
    ("Marco", "Verdi", [("Fisica", 22), ("Informatica", 29)])
  ]

# STEP 2: Creo il dizionario delle materie con i voti {Materia: [Voti]}
grades_dict = {} # Creo il dizionario vuoto da popolare
# Creo la lista di tuple (Materia, Voto)
for student in grades_list: # Eseguo il ciclo per ciascun elemento del dizionario
  subjects_list = student[2] # La lista di tuple si trova nella posizione 3 (indice 2)

  # Separo le materie dai voti all'interno della lista di tuple
  for subject in subjects_list: # Eseguo il ciclo per ciascun elemento del dizionario
    subject_name = subject[0] # Le materie sono nella posizione 1 (indice 0) di ciascuna tupla
    subject_grade = subject[1] # I voti sono nella posizione 2 (indice 1) di ciascuna tupla

    # Se una materia è già presente nel dizionario, creo la lista di voti
    if subject_name in grades_dict:
      grades_dict[subject_name].append(subject_grade) # Se una materia esiste già, allora aggiungo il voto
    else:
      grades_dict[subject_name] = [subject_grade] # Se una materia non esista, allora l'aggiungo al dizionario

print(f"Dizionario delle materie e dei voti: {grades_dict}\n") # Stampo il dizionario finale


# STEP 3: Per ogni corso, calcolo la media aritmetica dei voti (arrotondata a due decimali)
for subject, grades in grades_dict.items(): #Eseguo il calcolo per ciascun elemento del dizionario
  print(f"La media di {subject} è: {round(sum(grades)/len(grades), 2)}")


# STEP 4: Determino e stampo il corso con la media dei voti più alta
avg = 0 # Creo la variabile "media" e la inizializzo a 0
for subject, grades in grades_dict.items(): # Eseguo il calcolo per ciascun elemento del dizionario
  if sum(grades)/len(grades) > avg: # Confronto la media calcolata, con il valore medio precedente
    avg = sum(grades)/len(grades)
    best_subject = subject
print(f"\nCorso con la media più alta: {best_subject} (media: {round(avg, 2)})") # Stampo la materia con la media più alta

# STEP 5: Creo un nuovo set contenente i nomi unici degli studenti che hanno preso almeno un voto superiore o uguale a 28
top_students = set() # Creo un set vuoto da popolare
for student in grades_list: # Eseguo il ciclo per ciascun elemento del dizionario
  for student_grades in student[2]:
    if student_grades[1] >= 28:
      top_students.add(student[0])
      break
print(f"\nStudenti con almeno un voto >= 28: {top_students}")

# STEP 6: stampo il numero totale di studenti iscritti (input originale) e il numero di studenti con voti >= 28.
print(f"\nStudenti totali: {len(grades_list)}")
print(f"Studenti con almeno un voto >= 28: {len(top_students)}")
```

```python
# STEP 1: Dati annidati e accesso multicriterio:
# Dato un orario settimanale rappresentato come lista annidata,
# dove ogni sottolista contiene le materie studiate in ogni ora di un giorno
# (ad esempio, orario[2][3] è la materia della quarta ora del mercoledì),
# scrivo una funzione che dato un giorno e un'ora (entrambe come indice zero-based)
# restituisca la materia corrispondente.
# Gestisci anche il caso in cui la posizione sia vuota (ad esempio quando non c'è lezione).

# Definisco l'orario delle lezioni
time_table_list = [
    ["matematica", "italiano", "storia"],
    ["scienze", "inglese", "arte"],
    ["filosofia", "ed. fisica", "musica"],
    ["meccanica", "disegno", "officina"],
    ["inglese", "storia", "filosofia"]
    ]

# --------------------------------------------------------
# 1. GESTIONE DEL GIORNO
# --------------------------------------------------------

# Richiedo il giorno desiderato
input_day = int(input("Inserisci il giorno della settimana (da 1 = lunedì a 7 = domenica): "))

# Definisco la lista dei giorni della settimana
days_list= ["lunedì", "martedì", "mercoledì", "giovedì", "venerdì", "sabato", "domenica"]

# Definisco se il giorno inserito è valido:
# a. Variabile errore per il giorno
day_error = 0
# b. Ciclo semplice per l'analisi del giorno e stampa dell'avviso all'utente
if input_day < 1 or input_day > 7: # Giorno non valido
  day_error = 1
  print("Giorno non valido")
elif 6 <= input_day <= 7: # Giorno = sabato o domenica
  day_error = 1
  print(f"Di {days_list[input_day-1]} non ci sono lezioni")

# --------------------------------------------------------
# 2. GESTIONE DELL'ORA
# --------------------------------------------------------

# Solo se il giorno inserito è valido, richiedo l'ora
if day_error == 0:
  input_hour = int(input("Inserisci l'ora (da 1 a 3): "))
  # Controllo che l'ora sia valida (compresa tra 1 e 3):
  # a. Definisco la variabile errore per l'ora
  hour_error = 0
  #b. Ciclo semplice per l'analisi dell'ora e stampa dell'avviso all'utente
  if input_hour < 1 or input_hour > 3: # Ora non valida
    hour_error = 1
    print("Ora non valida")

# --------------------------------------------------------
# 3. STAMPA DELLA RICHIESTA
# --------------------------------------------------------

# Se non ci sono errori, stampo il risultato della ricerca giorno + ora
if day_error == 0 and hour_error == 0:
  print(f"La lezione dell'ora {input_hour} del {days_list[input_day-1]} è: {time_table_list[input_day-1][input_hour-1]}")

#==============================================================================

# STEP 2: Filtraggio e rimozione:
# Data una lista di stringhe che contiene alcune stringhe duplicate e altre con errori di battitura o maiuscole/minuscole miste
# (ad esempio, ['python', 'Python', 'java', 'JAVA', 'c++', 'c++', 'java']), scrivo una funzione che restituisca una nuova lista
# ordinata di stringhe uniche in minuscolo, rimuovendo duplicati ignorando il case originale.

# Definisco la lista iniziale e la stampo
input_list = ["python", "Python", "java", "JAVA", "c++", "c++", "java"]
print(f"\nLista iniziale: {input_list}")

# Definisco una funzione che restituisce una stringa, ordinata, unica, in minuscolo, senza duplicati, ingnorando il case iniziale
def clean_sort_names(input_list):

  # 1. Trasformo tutti gli elementi della lista in minuscole
  input_list = [element.lower() for element in input_list]

  # 2. Casting a Set per eliminare i duplicati
  input_set = set(input_list)

  # 3. Casting a Lista come richiesto e la ordino alfabeticamente
  input_list = list(input_set)
  input_list.sort()

  # 4. Restituisco la lista ripulita e ordinata
  return input_list

# Eseguo la funzione e stampo l'output
print(f"Lsita finale, in minuscolo, senza duplicati e ordinata alfabeticamente: {clean_sort_names(input_list)}")

#==============================================================================

# STEP 3: Dizionari annidati e modifica dinamica:
# Dato un dizionario che rappresenta la pagella di uno studente, con materie come chiavi e voti come valori, implemento una funzione che:
#  - incrementa di 1 punto tutti i voti inferiori a 6,
#  - aggiunge una nuova materia con voto iniziale (ad esempio 'Storia': 8),
#  - rimuove eventuali materie con voto uguale a 10.
#  - restituisce il dizionario modificato.

# Definisco la pagella iniziale e la stampo
grades_dict = {"Matematica": 5, "Italiano": 8, "Inglese": 10}
print(f"\nPagella iniziale: {grades_dict}")

# Incremento di un punto i valori inferiori a 6 e stampo la pagella
for key in grades_dict:
  if grades_dict[key] < 6:
    grades_dict[key] = grades_dict[key] + 1
print(f"Pagella con voti aumentati: {grades_dict}")

# Aggiungo una nuova materia con voto iniziale (Storia: 8) e stampo la pagella
grades_dict["Storia"] = 8
print(f"Pagella con la nuova materia: {grades_dict}")

# Rimuovo le materie con voto uguale a 10 e stampo la pagella
for key in list(grades_dict.keys()):
  if grades_dict[key] == 10:
    grades_dict.pop(key)
print(f"Pagella senza le materie con voto 10: {grades_dict}")

#==============================================================================

# STEP 4: Conversioni e cast tra collezioni:
# Data una tupla di nomi con possibili duplicati, converto la tupla in un set per rimuovere i duplicati,
# quindi la riconverto in lista ordinata alfabeticamente. Implemento la procedura come funzione.

# Definisco la tupla iniziale di nomi e la stampo
names_tuple = ("Anna", "Luca", "anna", "Luca", "Marco")
print(f"\nTupla iniziale di nomi: {names_tuple}")

# Definisco la funzione
def clean_sort_names(names_tuple):

  # Casting a lista per poter modificare gli elementi
  names_list = list(names_tuple)

  # 1. Mi assicuro che i nomi siano scritti nello stesso modo:
  # a. Tutto in minuscole
  names_lower_case_list = [element.lower() for element in names_list]
  # b. Capitalizzo la prima lettera di ogni elemento
  names_capitalized_list = [element.capitalize() for element in names_lower_case_list]

  # 2. Casting a Set per eliminare i duplicati
  names_set = set(names_capitalized_list)

  # 3. Casting a Lista come richiesto, la ordino alfabeticamente e la stampo
  names_list = list(names_set)
  names_list.sort()

  # 4. Restituisco la lista ripulita e senza duplicati
  return names_list

# Eseguo la funzione e stampo l'output
print(f"Lista finale, in minuscolo, senza duplicati e ordinata alfabeticamente: {clean_sort_names(names_tuple)}")

#==============================================================================

# STEP 5: Copia profonda di liste annidate:
# Data una lista annidata (lista di liste), scrivo una funzione che ne crea una copia indipendente (deep copy)
# senza usare funzioni esterne. Dimostro che modificando un elemento della lista originale non si modifica la copia.

# Definisco la lista annidata iniziale e la stampo
input_list = [[1, 2], [3, 4]]
print(f"\nLista annidata iniziale: {input_list}")

# Creo una funzione che esegue il deep copy della lista iniziale senza usare funzioni esterne
def deep_copy(input_list):

  # 1. Creo una lista vuota da popolare
  deep_copy_list = []

  # 2. Analizzo ogni elemento della lista e lo copio in una nuova lista
  deep_copy_list = [element.copy() for element in input_list]

  # 3. Ritorno la lista copiata
  return deep_copy_list

# Eseguo la funzione e stampo il risultato
deep_copy_list = deep_copy(input_list)
print(f"Lista annidata deep copy: {deep_copy_list}")

# Modifico un elemento della lista copiata e stampo entrambe le liste
deep_copy_list[0][1] = 99
print(f"Lista annidata deep copy modificata: {deep_copy_list}")
print(f"Lista annidata iniziale non modificata: {input_list}")

```
